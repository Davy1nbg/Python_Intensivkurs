### 1\. Einführung in die Objektorientierte Programmierung (OOP) (10 Min.)

**Was ist OOP?**

Stell dir vor, du baust mit LEGO®. Anstatt jeden einzelnen Stein immer wieder neu zu formen, hast du vorgefertigte Bausteine (z.B. einen roten 2x4-Stein, ein Fenster, ein Rad). OOP ist ein Programmierparadigma, das ähnlich funktioniert. Es geht darum, Software aus "Objekten" zu erstellen, die Daten und Funktionen (Methoden genannt) bündeln, die auf diese Daten wirken.

**Warum OOP?**

  * **Wiederverwendbarkeit:** Einmal definierte "Baupläne" (Klassen) können immer wieder verwendet werden, um neue "Bausteine" (Objekte) zu erstellen.
  * **Modularität:** Code wird in logische Einheiten (Objekte) aufgeteilt. Das macht Programme übersichtlicher und einfacher zu warten. Wenn ein Teil geändert werden muss, beeinflusst das nicht unbedingt den Rest des Programms.
  * **Abstraktion:** Komplexe Vorgänge können hinter einfachen Schnittstellen versteckt werden. Du musst nicht wissen, *wie* ein Auto intern funktioniert, um es fahren zu können.
  * **Bessere Modellierung der realen Welt:** Viele Dinge in der realen Welt lassen sich gut als Objekte mit Eigenschaften und Verhaltensweisen beschreiben (z.B. ein Student hat einen Namen, eine Matrikelnummer und kann sich für Kurse einschreiben).

**Kernkonzepte der OOP (heute im Fokus):**

  * **Klassen (Classes):** Die "Baupläne" oder "Schablonen" für Objekte.
  * **Objekte (Objects/Instances):** Konkrete Ausprägungen oder "Instanzen" einer Klasse.
  * **Methoden (Methods):** Funktionen, die zu einer Klasse gehören und das Verhalten von Objekten definieren.
  * **Attribute (Attributes):** Variablen, die zu einer Klasse gehören und die Eigenschaften von Objekten speichern.

### 2\. Klassen: Die Blaupause blueprint (20 Min.)

Eine **Klasse** ist wie ein Bauplan für Häuser. Der Bauplan selbst ist kein Haus, aber er beschreibt, wie ein Haus gebaut wird (Anzahl der Zimmer, Fenster, Türen etc.).

In Python definieren wir eine Klasse mit dem Schlüsselwort `class`.

```
class Student:
    pass # pass bedeutet hier "vorerst leer"
```

**Der Konstruktor: `__init__()`**

Die speziellste Methode in einer Klasse ist oft der **Konstruktor**. In Python heißt diese Methode `__init__()`. Sie wird automatisch aufgerufen, wenn ein neues Objekt dieser Klasse erstellt wird. Hier werden typischerweise die Anfangswerte für die Eigenschaften (Attribute) eines Objekts festgelegt.

  * Das erste Argument jeder Methode in einer Klasse (auch des Konstruktors) ist immer `self`.
  * `self` repräsentiert die konkrete Instanz (das Objekt), mit der gerade gearbeitet wird.

<!-- end list -->

```
class Student:
    # Der Konstruktor
    def __init__(self, name, matrikelnummer):
        # Attribute der Instanz (des Objekts)
        self.name = name
        self.matrikelnummer = matrikelnummer
        self.kurse = [] # Jeder Student startet mit einer leeren Liste von Kursen

        print(f"Student {self.name} mit Matrikelnr. {self.matrikelnummer} wurde erstellt!")

# Erläuterung:
# - `class Student:`: Definiert eine neue Klasse namens "Student".
# - `def __init__(self, name, matrikelnummer):`: Der Konstruktor.
#   - `self`: Bezieht sich auf das gerade erstellte Objekt.
#   - `name`, `matrikelnummer`: Parameter, die beim Erstellen eines Studentenobjekts übergeben werden.
# - `self.name = name`: Erstellt ein Attribut "name" für das Objekt und weist ihm den übergebenen Wert zu.
# - `self.matrikelnummer = matrikelnummer`: Erstellt ein Attribut "matrikelnummer".
# - `self.kurse = []`: Erstellt ein Attribut "kurse" als leere Liste.
```

**Klassenattribute vs. Instanzattribute**

  * **Instanzattribute:** Gehören zu einer spezifischen Instanz/Objekt (z.B. `self.name`). Jeder Student hat seinen *eigenen* Namen. Sie werden meist im `__init__` definiert.
  * **Klassenattribute:** Werden von *allen* Instanzen einer Klasse geteilt. Sie werden direkt innerhalb der Klasse, aber außerhalb von Methoden definiert.

<!-- end list -->

```
class Student:
    # Klassenattribut
    universitaet = "Technische Hochschule Beispielhausen"

    def __init__(self, name, matrikelnummer):
        # Instanzattribute
        self.name = name
        self.matrikelnummer = matrikelnummer
        self.kurse = []
```

### 3\. Objekte: Die Instanzen 🏠 (20 Min.)

Ein **Objekt** ist eine konkrete Ausprägung (Instanz) einer Klasse. Wenn die Klasse `Student` der Bauplan ist, dann ist jeder einzelne Student (z.B. "Max Mustermann", "Erika Musterfrau") ein Objekt dieser Klasse.

**Objekte erstellen (Instanziierung):**

Um ein Objekt zu erstellen, rufen wir die Klasse wie eine Funktion auf und übergeben die Argumente, die der `__init__`-Methode erwartet (außer `self`, das wird automatisch übergeben).

```
# Erstellen von Objekten (Instanzen) der Klasse Student
student1 = Student("Max Mustermann", "12345")
student2 = Student("Erika Musterfrau", "67890")

# student1 und student2 sind jetzt zwei unabhängige Objekte.
```

**Zugriff auf Attribute:**

Wir können auf die Attribute eines Objekts mit der Punkt-Notation zugreifen: `objektname.attributname`.

```
print(f"Name von Student 1: {student1.name}") # Ausgabe: Max Mustermann
print(f"Matrikelnummer von Student 2: {student2.matrikelnummer}") # Ausgabe: 67890
print(f"Universität von Student 1: {student1.universitaet}") # Ausgabe: Technische Hochschule Beispielhausen
print(f"Universität von Student 2: {student2.universitaet}") # Ausgabe: Technische Hochschule Beispielhausen

# Ändern eines Instanzattributs
student1.name = "Maximilian Mustermann"
print(f"Neuer Name von Student 1: {student1.name}") # Ausgabe: Maximilian Mustermann

# Das Klassenattribut kann auch über die Klasse direkt oder ein Objekt zugegriffen werden
print(Student.universitaet) # Ausgabe: Technische Hochschule Beispielhausen
```

Jedes Objekt hat seine eigenen Werte für Instanzattribute, teilt sich aber die Klassenattribute.

```
print(student1.kurse) # Ausgabe: []
student1.kurse.append("Programmieren 1")
print(student1.kurse) # Ausgabe: ['Programmieren 1']
print(student2.kurse) # Ausgabe: [] - Die Kurse von student2 sind davon unberührt
```

### 4\. Methoden: Was Objekte können 🛠️ (20 Min.)

**Methoden** sind Funktionen, die innerhalb einer Klasse definiert werden. Sie bestimmen, was Objekte dieser Klasse "tun" können oder wie sie sich verhalten.

  * Das erste Argument einer Instanzmethode ist immer `self`, das auf das Objekt verweist, das die Methode aufruft.

<!-- end list -->

```
class Student:
    universitaet = "Technische Hochschule Beispielhausen"

    def __init__(self, name, matrikelnummer):
        self.name = name
        self.matrikelnummer = matrikelnummer
        self.kurse = []

    # Eine Instanzmethode, um einen Kurs hinzuzufügen
    def kurs_einschreiben(self, kursname):
        if kursname not in self.kurse:
            self.kurse.append(kursname)
            print(f"{self.name} hat sich für {kursname} eingeschrieben.")
        else:
            print(f"{self.name} ist bereits für {kursname} eingeschrieben.")

    # Eine Instanzmethode, um Informationen über den Studenten auszugeben
    def zeige_info(self):
        print("--- Studenteninfo ---")
        print(f"Name: {self.name}")
        print(f"Matrikelnummer: {self.matrikelnummer}")
        print(f"Universität: {self.universitaet}") # Zugriff auf Klassenattribut
        print(f"Eingeschriebene Kurse: {', '.join(self.kurse) if self.kurse else 'Keine'}")
```

**Methoden aufrufen:**

Methoden werden ähnlich wie Attribute über die Punkt-Notation aufgerufen: `objektname.methodenname(argumente)`.

```
# Objekte erstellen
student1 = Student("Max Mustermann", "12345")
student2 = Student("Erika Musterfrau", "67890")

# Methoden aufrufen
student1.kurs_einschreiben("Mathematik 1")
student1.kurs_einschreiben("Python Grundlagen")
student1.kurs_einschreiben("Mathematik 1") # Bereits eingeschrieben

student2.kurs_einschreiben("Statistik")

print("\n--- Aufruf von zeige_info ---")
student1.zeige_info()
# Ausgabe:
# --- Studenteninfo ---
# Name: Max Mustermann
# Matrikelnummer: 12345
# Universität: Technische Hochschule Beispielhausen
# Eingeschriebene Kurse: Mathematik 1, Python Grundlagen

print() # Leerzeile für bessere Lesbarkeit
student2.zeige_info()
# Ausgabe:
# --- Studenteninfo ---
# Name: Erika Musterfrau
# Matrikelnummer: 67890
# Universität: Technische Hochschule Beispielhausen
# Eingeschriebene Kurse: Statistik
```
