### 1. Datenpersistenz und SQLite

Bisher gingen alle Programmdaten beim Schließen der Applikation verloren. Um Informationen dauerhaft zu sichern, nutzen wir Datenbanken. **SQLite** zeichnet sich dadurch aus, dass die gesamte Datenbank in einer lokalen Datei gespeichert wird, was sie ideal für Desktop-Anwendungen und wissenschaftliche Prototypen macht.

**Vorteile von SQLite in Python:**

* **Serverlos:** Keine Installation eines Datenbank-Servers notwendig.
* **Standardbibliothek:** Das Modul `sqlite3` ist bereits in Python enthalten.
* **ACID-Konformität:** Gewährleistet Datensicherheit und Integrität.

---

### 2. Datenbank-Interaktion via Python

Die Kommunikation mit der Datenbank folgt einem festen Protokoll: Verbindung aufbauen, Cursor erstellen, Befehl ausführen, Änderungen bestätigen (Commit) und Verbindung schließen.

```
import sqlite3

def initialisiere_datenbank():
    # Verbindung zur Datenbankdatei herstellen (wird erstellt, falls nicht existent)
    connection = sqlite3.connect("anwendung.db")
    cursor = connection.cursor()

    # Tabelle erstellen
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS nutzer (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            email TEXT UNIQUE
        )
    ''')

    connection.commit()
    connection.close()

initialisiere_datenbank()
```

### 3. Integration: GUI trifft Datenbank

Die Verknüpfung erfolgt über Callback-Funktionen. Die GUI sammelt die Benutzereingaben, und die Logik-Funktion schreibt diese in die Datenbank.

```
import tkinter as tk
from tkinter import messagebox

def datensatz_speichern():
    name = entry_name.get()
    mail = entry_mail.get()

    if name and mail:
        conn = sqlite3.connect("anwendung.db")
        cur = conn.cursor()
        try:
            cur.execute("INSERT INTO nutzer (name, email) VALUES (?, ?)", (name, mail))
            conn.commit()
            messagebox.showinfo("Erfolg", "Daten wurden persistent gespeichert.")
        except sqlite3.IntegrityError:
            messagebox.showerror("Fehler", "Diese E-Mail-Adresse existiert bereits.")
        finally:
            conn.close()
    else:
        messagebox.showwarning("Eingabe fehlt", "Bitte füllen Sie alle Felder aus.")

# GUI Setup
root = tk.Tk()
root.title("Datenbank-Interface")

tk.Label(root, text="Name:").grid(row=0, column=0, padx=10, pady=5)
entry_name = tk.Entry(root)
entry_name.grid(row=0, column=1, padx=10, pady=5)

tk.Label(root, text="E-Mail:").grid(row=1, column=0, padx=10, pady=5)
entry_mail = tk.Entry(root)
entry_mail.grid(row=1, column=1, padx=10, pady=5)

btn_save = tk.Button(root, text="In DB speichern", command=datensatz_speichern)
btn_save.grid(row=2, column=0, columnspan=2, pady=10)

root.mainloop()
```
