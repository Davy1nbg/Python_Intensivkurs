### 1. Paradigma der Objektorientierung

Objektorientierte Programmierung (OOP) ist ein Konzept, bei dem Software als eine Sammlung von interagierenden Objekten modelliert wird. Ein Objekt bündelt **Daten** (Attribute) und **Verhalten** (Methoden).

**Vorteile der OOP:**

* **Abstraktion:** Reduktion komplexer Realitäten auf wesentliche Merkmale.
* **Modularität:** Kapselung von Logik in eigenständige Einheiten zur Verbesserung der Wartbarkeit.
* **Wiederverwendbarkeit:** Effiziente Nutzung definierter Klassen für eine Vielzahl von Instanzen.

---

### 2. Klassen: Definition und Konstruktion

Eine Klasse fungiert als Schablone oder Bauplan für Objekte.

#### Der Konstruktor `__init__`

Die Methode `__init__` ist eine spezielle Methode, die bei der Erstellung (Instanziierung) eines Objekts automatisch aufgerufen wird. Sie dient der Initialisierung des Objektzustands.

* **`self`:** Dieser Parameter referenziert die aktuelle Instanz der Klasse und ermöglicht den Zugriff auf deren Attribute und Methoden.

```
class Student:
    # Der Konstruktor initialisiert die Instanzattribute
    def __init__(self, name, matrikelnummer):
        self.name = name
        self.matrikelnummer = matrikelnummer
        self.kurse = []  # Initialisierung einer leeren Liste
```

#### Instanz- vs. Klassenattribute

* **Instanzattribute:** Sind spezifisch für jedes Objekt (z.B. der Name eines Studenten).
* **Klassenattribute:** Werden global für alle Instanzen einer Klasse definiert und von diesen geteilt (z.B. der Name der Universität).

---

### 3. Instanziierung und Attributzugriff

Ein Objekt ist eine konkrete Ausprägung einer Klasse. Der Prozess der Erstellung wird als Instanziierung bezeichnet.

```
# Instanziierung zweier unabhängiger Objekte
student1 = Student("Max Mustermann", "12345")
student2 = Student("Erika Musterfrau", "67890")

# Zugriff via Punkt-Notation
print(student1.name)
```

---

### 4. Methoden: Funktionalität von Objekten

Methoden sind innerhalb einer Klasse definierte Funktionen, die das Verhalten der Objekte festlegen.

```
class Student:
    universitaet = "Uni Mannheim" # Klassenattribut

    def __init__(self, name, matrikelnummer):
        self.name = name
        self.matrikelnummer = matrikelnummer
        self.kurse = []

    def kurs_einschreiben(self, kursname):
        if kursname not in self.kurse:
            self.kurse.append(kursname)
            print(f"Einschreibung für {kursname} erfolgreich.")

    def zeige_info(self):
        print(f"Student: {self.name} | Universität: {self.universitaet}")
```
