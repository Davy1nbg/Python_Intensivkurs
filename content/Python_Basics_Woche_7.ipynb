````python
Absolut! Dieses Kursmaterial für Block 2 ist bereits sehr gut strukturiert und detailliert. Ich habe es für die Verwendung in einem Jupyter Notebook optimiert, indem ich es in passende Zellen aufgeteilt und sichergestellt habe, dass Befehle klar als Code-Blöcke formatiert sind. Die Erklärungen und Analogien wurden beibehalten.

Hier ist das optimierte Kursmaterial für Block 2:

# Block 2: Was ist Django? Installation (ca. 60-75 Min.) 🐍

**Ziel:** Die Studenten verstehen, was ein Webframework ist, warum Django eine gute Wahl für Python-Entwickler ist, und sie können Django in einer virtuellen Umgebung installieren, ein erstes Projekt anlegen und den Entwicklungsserver starten.

## 1. Was ist ein Webframework? Warum Django? (ca. 15 Minuten)

**Anknüpfung an Block 1:**
"Wir haben in Block 1 gesehen, dass Webserver HTML-Seiten an Browser schicken. Man könnte diese HTML-Seiten für eine kleine Webseite von Hand schreiben. Aber stellt euch vor, ihr habt eine große Webseite mit hunderten Unterseiten, Benutzeranmeldungen, Suchfunktionen oder Inhalten, die sich ständig ändern (wie Nachrichten oder Produkte in einem Shop). Das alles manuell mit einzelnen HTML-Dateien und der Verarbeitung von Anfragen zu managen, wäre unglaublich aufwendig und fehleranfällig!"

**Die Herausforderung der Webentwicklung:**
* Wie reagiert man auf verschiedene URLs (z.B. `/home`, `/artikel/42`, `/kontakt`)?
* Wie holt man Daten aus einer Datenbank und zeigt sie an?
* Wie verarbeitet man Formulareingaben von Nutzern?
* Wie stellt man sicher, dass alles sicher ist?
* Wie organisiert man seinen Code übersichtlich?


**Die Lösung: Webframeworks!**

* **Was ist ein Webframework?**
    * Ein Webframework ist wie ein **hochspezialisierter Werkzeugkasten** oder ein umfangreiches Set an **Kochrezepten und Küchenhelfern** für Webentwickler.
    * Es nimmt einem viele immer wiederkehrende, grundlegende Aufgaben bei der Webentwicklung ab.
    * Es gibt eine Struktur und Regeln vor, wie man Webanwendungen baut.
    * **Ziel:** Entwickler müssen nicht jedes Mal das Rad neu erfinden, sondern können sich auf die einzigartigen Funktionen ihrer Webseite konzentrieren.

* **Analogie: Fertighaus vs. Selbst Holz schlagen** 🏡
    * **Ohne Framework:** Man muss jeden Baum selbst fällen, jedes Brett sägen, jeden Nagel selbst schmieden, um ein Haus zu bauen. Sehr mühsam!
    * **Mit Framework:** Man bekommt ein Fertighaus-System. Das Fundament, die Grundmauern, Standardfenster und -türen sind schon da oder leicht einzusetzen. Man kümmert sich dann um die Inneneinrichtung und spezielle Anpassungen.


**Warum ausgerechnet Django?**

* **In Python geschrieben:** Da ihr bereits Python-Grundlagen habt, ist der Einstieg oft leichter. Ihr könnt euer vorhandenes Wissen nutzen!
* **"Batteries Included" (Batterien enthalten):** Django kommt mit sehr vielen eingebauten Funktionen, die man oft braucht:
    * Eine Administrationsoberfläche (wie ein Backend für Seitenbetreiber).
    * Werkzeuge für die einfache Arbeit mit Datenbanken (ORM - Object-Relational Mapper).
    * Ein System für Templates (um HTML dynamisch zu erzeugen).
    * Sicherheitsfunktionen.
    * *(Hinweis an Studenten: Vieles davon werden wir heute nicht im Detail sehen, aber es ist gut zu wissen, dass es da ist.)*
* **Sehr verbreitet und etabliert:** Große Webseiten wie Instagram, Spotify (teilweise), Pinterest oder Mozilla nutzen Django. Das bedeutet:
    * Gute Dokumentation.
    * Eine riesige Community, bei der man Hilfe findet.
    * Viele Erweiterungen und Pakete.
* **Skalierbar:** Eignet sich für kleine Hobbyprojekte genauso wie für sehr große, komplexe Anwendungen.

**Was macht Django konkret (vereinfacht für unser heutiges Ziel)?**
* Django hilft uns, auf Anfragen (Requests) vom Browser zu reagieren.
* Es ermöglicht uns, Python-Code auszuführen, um auf diese Anfragen zu reagieren (z.B. etwas zu berechnen, aus einer Datei zu lesen).
* Am Ende hilft es uns, dynamisch HTML (als Teil der Response) zu erzeugen und an den Browser zurückzuschicken.


## 2. Vorbereitung: Virtuelle Umgebung (ca. 15 Minuten)

**Das Problem: Abhängigkeits-Chaos** 얽힌 실타래 (verhedderter Fadenknäuel)
Stellt euch vor, ihr arbeitet an mehreren Python-Projekten gleichzeitig:
* Projekt A braucht eine bestimmte Version von Paket X (z.B. Version 1.0).
* Projekt B braucht aber eine neuere Version von Paket X (z.B. Version 2.0).
Wenn ihr alle Pakete global (also für euren ganzen Computer) installiert, kann das zu Konflikten führen. Ein Projekt funktioniert dann vielleicht nicht mehr richtig, weil ein anderes Projekt eine Abhängigkeit geändert hat.

**Die Lösung: Virtuelle Umgebungen (venv) – Für jedes Projekt eine eigene saubere Werkstatt!** 🛠️
* Eine virtuelle Umgebung ist ein **isolierter Ordner** für jedes eurer Python-Projekte.
* In dieser Umgebung werden Python selbst (bzw. eine Kopie davon) und alle benötigten Pakete (wie Django) **nur für dieses eine Projekt** installiert.
* **Vorteile:**
    * Projekte beeinflussen sich nicht gegenseitig.
    * Man kann für jedes Projekt genau die Paketversionen verwenden, die es braucht.
    * Hält das globale Python-System sauber.


**Schritt-für-Schritt: Virtuelle Umgebung erstellen (Live-Coding)**

1.  **Projektordner anlegen:**
    * Öffnet eure Kommandozeile (Terminal, PowerShell, Git Bash etc.).
    * Erstellt einen neuen Ordner für unser heutiges Django-Projekt, z.B. `django_ein_stieg`.

    ```bash
    mkdir django_ein_stieg
    ```
    * Wechselt in diesen Ordner:
    ```bash
    cd django_ein_stieg
    ```

2.  **Virtuelle Umgebung erstellen:**
    * Der Befehl dazu lautet: `python -m venv <name_der_umgebung>`
    * Wir nennen unsere Umgebung einfach `venv` (das ist eine häufige Konvention):

    ```bash
    python -m venv venv
    ```
    * `python -m venv`: Sagt Python, es soll sein eingebautes Modul `venv` ausführen.
    * `venv` (der zweite Teil): Ist der Name des Ordners, der für die virtuelle Umgebung erstellt wird.
    * *Schaut euch den Inhalt eures Projektordners an. Es sollte jetzt einen Unterordner `venv` geben. Darin befinden sich verschiedene Dateien und Ordner, die Python für diese isolierte Umgebung benötigt. (Nicht im Detail darauf eingehen).*

3.  **Virtuelle Umgebung aktivieren:**
    * **WICHTIG:** Bevor ihr Pakete installiert oder mit eurem Projekt arbeitet, müsst ihr die virtuelle Umgebung "aktivieren". Das sagt eurer Kommandozeile, dass sie ab jetzt das Python und die Pakete aus diesem `venv`-Ordner verwenden soll.
    * **Windows** (in der Kommandozeile oder PowerShell):
        ```bash
        .\venv\Scripts\activate
        ```
    * **macOS / Linux** (im Terminal):
        ```bash
        source venv/bin/activate
        ```
    * **Erkennungszeichen:** Nach der Aktivierung ändert sich oft der Prompt (die Zeile, in der ihr Befehle eingebt). Häufig wird der Name der virtuellen Umgebung in Klammern vorangestellt, z.B. `(venv) C:\Users\DeinName\django_ein_stieg>`.
    * *Hinweis: Um die virtuelle Umgebung später wieder zu deaktivieren (wenn ihr mit der Arbeit am Projekt fertig seid), tippt einfach `deactivate` in die Kommandozeile.*


## 3. Django installieren (ca. 15 Minuten)

**Voraussetzung:** Die virtuelle Umgebung (`venv`) muss **aktiviert** sein! (Prompt sollte `(venv)` o.ä. anzeigen).

**PIP – Der Paketmanager für Python 📦**
* `pip` ist das Standardwerkzeug in Python, um externe Bibliotheken und Pakete zu installieren (und zu verwalten). Django ist so ein Paket.
* `pip` wurde automatisch mit Python installiert und ist auch in eurer virtuellen Umgebung verfügbar.
* **Analogie:** `pip` ist wie ein App Store oder ein Paketdienst speziell für Python-Software.


**Django installieren (Live-Coding):**
* Gebt in eurer aktivierten Kommandozeile folgenden Befehl ein:

    ```bash
    pip install django
    ```
* *Beobachtet die Ausgabe: `pip` wird nun Django und alle Pakete, von denen Django abhängt (sog. "dependencies"), aus dem Internet herunterladen und in eurer virtuellen Umgebung (`venv/Lib/site-packages`) installieren. Das kann einen Moment dauern.*


**Installation überprüfen (optional, aber empfehlenswert):**

1.  **`pip freeze`**: Dieser Befehl zeigt euch alle Pakete, die aktuell in eurer aktiven virtuellen Umgebung installiert sind, inklusive ihrer Versionen.
    ```bash
    pip freeze
    ```
    Ihr solltet `Django==X.Y.Z` (mit der aktuellen Versionsnummer) in der Liste sehen.

2.  **`django-admin --version`**: Django bringt ein Kommandozeilenwerkzeug namens `django-admin` mit. Damit können wir die Version prüfen:
    ```bash
    django-admin --version
    ```
    Dies sollte die installierte Django-Version ausgeben.

**Erfolg!** Django ist nun in eurer isolierten Projektumgebung einsatzbereit. ✅


## 4. Ein Django-Projekt starten (ca. 15-25 Minuten)

**Was ist ein Django-Projekt? 🤔**
* Ein Django-Projekt ist der **Container oder die Hülle** für eine komplette Webanwendung oder Webseite.
* Es enthält alle Konfigurationen (z.B. Datenbankeinstellungen, welche Django-Apps verwendet werden) und kann eine oder mehrere "Apps" beinhalten (dazu mehr in Block 3).
* **Analogie:** Stellt euch ein Haus vor. Das Django-Projekt ist das gesamte **Haus** mit Fundament, Dach und allen Anschlüssen. Später könnt ihr in diesem Haus verschiedene **Zimmer (Apps)** einrichten, z.B. ein Wohnzimmer, eine Küche etc.


**Django-Projekt erstellen (Live-Coding):**

* **WICHTIG:** Stellt sicher, dass ihr euch in eurem Haupt-Projektordner befindet (also z.B. `django_ein_stieg`), **NICHT** im `venv`-Ordner! Eure virtuelle Umgebung muss aber weiterhin aktiviert sein.
* Der Befehl zum Erstellen eines Projekts lautet: `django-admin startproject <projektname> [optional: zielordner]`
* Wir nennen unser Projekt `mein_erstes_projekt` und wollen, dass die Projektdateien direkt in unserem aktuellen Ordner (`django_ein_stieg`) erstellt werden (und nicht in einem zusätzlichen Unterordner). Dafür fügen wir am Ende einen Punkt `.` hinzu:

    ```bash
    django-admin startproject mein_erstes_projekt .
    ```
    * `django-admin`: Das Kommandozeilen-Tool von Django.
    * `startproject`: Der Befehl zum Erstellen eines neuen Projekts.
    * `mein_erstes_projekt`: Der Name, den wir unserem Projekt geben.
    * `.` (der Punkt am Ende): **Sehr wichtig!** Er sagt Django: "Erstelle die Projektdateien im aktuellen Verzeichnis." Lässt man ihn weg, würde Django einen zusätzlichen Ordner `mein_erstes_projekt` anlegen, was zu einer verschachtelten Struktur wie `django_ein_stieg/mein_erstes_projekt/mein_erstes_projekt/` führen kann, die wir vermeiden wollen.

**Die Ordnerstruktur eines Django-Projekts (kurzer Überblick, nicht jedes Detail!):**

Schaut euch nun den Inhalt eures `django_ein_stieg`-Ordners an. Ihr seht dort:
* `manage.py`:
    * Das Schweizer Taschenmesser für euer Django-Projekt! Eine sehr wichtige Datei. Mit ihr werdet ihr viele Django-spezifische Befehle ausführen, z.B. den Entwicklungsserver starten, Datenbankänderungen vornehmen oder neue "Apps" erstellen.
* `mein_erstes_projekt/` (ein Unterordner mit dem Namen eures Projekts):
    * Dieser Ordner enthält die eigentlichen Konfigurationsdateien für euer Projekt:
        * `__init__.py`: Eine leere Datei, die Python sagt, dass dieser Ordner als Paket behandelt werden soll.
        * `settings.py`: Die Haupt-Einstellungsdatei eures Projekts. Hier stehen Dinge wie Datenbankkonfiguration, installierte Apps, Zeitzone, Verweise auf statische Dateien und Templates etc. (Heute nur kurz erwähnen, wir schauen da später mal rein.)
        * `urls.py`: Die "Wegweiser"-Datei. Hier wird festgelegt, welche URL-Pfade zu welchen Teilen eurer Django-Anwendung führen (also welche Funktion aufgerufen wird, wenn ein Nutzer eine bestimmte URL eingibt). (Kommt in Block 3)
        * `asgi.py` & `wsgi.py`: Dateien, die für das Deployment (Bereitstellen der Webseite auf einem echten Webserver) wichtig sind. (Für heute ignorieren wir diese.)

**Der Entwicklungsserver – Unsere erste Django-Webseite live sehen! 🚀**

* Django kommt mit einem eingebauten, leichtgewichtigen Webserver. Dieser ist **nur für die Entwicklung gedacht**, nicht für den Live-Betrieb einer echten Webseite, da er nicht sehr performant oder sicher ist. Aber zum Testen und Entwickeln ist er perfekt!

**Server starten (Live-Coding):**
* Stellt sicher, dass eure virtuelle Umgebung aktiv ist und ihr euch im Hauptordner eures Projekts befindet (dort, wo die `manage.py` Datei liegt).
* Gebt folgenden Befehl ein:

    ```bash
    python manage.py runserver
    ```


**Was passiert im Terminal?**
* Ihr seht einige Zeilen Output. Wenn alles gut geht, steht dort so etwas wie:
    ```
    Performing system checks...

    System check identified no issues (0 silenced).
    May 26, 2025 - 08:00:00  # Datum und Uhrzeit werden aktuell sein
    Django version 4.2.X, using settings 'mein_erstes_projekt.settings' # Version kann variieren
    Starting development server at http://127.0.0.1:8000/
    Quit the server with CTRL-BREAK (Windows) or CTRL-C (Ctrl+C).
    ```
* `http://127.0.0.1:8000/`: Das ist die Adresse, unter der euer Entwicklungsserver jetzt läuft!
    * `127.0.0.1` (oder `localhost`): Eine spezielle IP-Adresse, die immer auf euren eigenen Computer verweist.
    * `:8000`: Die Portnummer. Ein Computer kann über verschiedene Ports unterschiedliche Dienste anbieten. Django nutzt standardmäßig Port 8000.


**Im Browser öffnen:**
* Öffnet jetzt einen Webbrowser eurer Wahl (Chrome, Firefox etc.).
* Gebt die Adresse `http://127.0.0.1:8000/` in die Adresszeile ein und drückt Enter.

🎉 **ERFOLG!** 🎉 Ihr solltet die Django-Willkommensseite sehen ("The install worked successfully! Congratulations!"). Diese Seite mit der Rakete wird von Django automatisch generiert.

*Betont, dass dies ein wichtiger Meilenstein ist: Die Studenten haben einen Webserver auf ihrem eigenen Rechner gestartet, der eine dynamisch generierte Seite ausliefert!*


**Server stoppen:**
* Um den Entwicklungsserver wieder zu beenden, wechselt zurück zu eurem Kommandozeilenfenster (wo der Server läuft) und drückt die Tastenkombination `Strg + C` (Windows) bzw. `Ctrl + C` (macOS/Linux).


**Zwischenziel-Check:**
* Haben alle Django installiert, ein Projekt erstellt und die Willkommensseite gesehen?
* Gibt es Fragen zur Ordnerstruktur oder zum Entwicklungsserver?

````
