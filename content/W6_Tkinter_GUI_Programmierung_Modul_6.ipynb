## Modul 6: Grafische Benutzeroberflächen (GUI) mit Tkinter I

### Lernziele

* Verständnis des Paradigmenwechsels von sequenzieller zu **ereignisgesteuerter Programmierung**.
* Beherrschung der Kernarchitektur einer GUI: Hauptfenster (Root), Widgets und der **Event-Loop**.
* Anwendung von Layout-Managern zur Strukturierung von Oberflächen (`pack` und `grid`).

---

### 1. Grundlagen: Ereignisgesteuerte Programmierung

Im Gegensatz zu CLI-Skripten, die linear ablaufen, agieren GUI-Programme reaktiv. Nach der Initialisierung tritt das Programm in den **Mainloop** ein – eine Endlosschleife, die auf Benutzereingaben (Events) wartet.

**Der Event-Loop Prozess:**

1. Initialisierung der grafischen Komponenten.
2. Eintritt in den Wartezustand (`mainloop`).
3. Erkennung eines Events (z. B. Mausklick, Tastatureingabe).
4. Ausführung der verknüpften Callback-Funktion.
5. Aktualisierung der Anzeige und Rückkehr in den Wartezustand.

---

### 2. Anatomie einer Tkinter-Applikation

Eine Tkinter-Anwendung folgt einem hierarchischen Aufbau: Das Hauptfenster fungiert als Container für verschiedene Steuerelemente (Widgets).

```
import tkinter as tk
from tkinter import messagebox

# Logik-Bereich (Callback-Funktionen)
def verarbeite_eingabe():
    nutzer_name = eingabefeld.get()
    if nutzer_name:
        label_ausgabe.config(text=f"Willkommen, {nutzer_name}", fg="navy")
    else:
        messagebox.showwarning("Eingabefehler", "Das Feld darf nicht leer sein.")

# GUI-Setup
root = tk.Tk()
root.title("Anwendungsfenster Modul 6")
root.geometry("400x300")

# Widget-Instanziierung
instruktion = tk.Label(root, text="Bitte Identität bestätigen:", font=("Helvetica", 10))
eingabefeld = tk.Entry(root)
button_bestaetigen = tk.Button(root, text="Validieren", command=verarbeite_eingabe)
label_ausgabe = tk.Label(root, text="", font=("Helvetica", 10, "bold"))

# Layout-Management via .pack()
instruktion.pack(pady=10)
eingabefeld.pack(pady=5)
button_bestaetigen.pack(pady=10)
label_ausgabe.pack(pady=20)

# Start des Event-Loops
root.mainloop()
```

---

### 3. Layout-Management: Das Grid-System

Während `.pack()` Widgets lediglich linear stapelt, ermöglicht der `grid`-Manager eine tabellarische Anordnung über Zeilen (`row`) und Spalten (`column`). Dies ist der Standard für komplexe Formulare.

**Grid-Parameter:**

* `sticky`: Ausrichtung des Widgets innerhalb der Zelle (N, S, E, W).
* `columnspan`: Erstreckung eines Widgets über mehrere Spalten.
* `padx` / `pady`: Außenabstände der Zellen.

```
# Beispiel eines strukturierten Logins
tk.Label(root, text="Benutzerkennung:").grid(row=0, column=0, sticky="w", padx=5)
tk.Entry(root).grid(row=0, column=1, padx=5, pady=5)

tk.Label(root, text="Kennwort:").grid(row=1, column=0, sticky="w", padx=5)
tk.Entry(root, show="*").grid(row=1, column=1, padx=5, pady=5)

tk.Button(root, text="Login").grid(row=2, column=0, columnspan=2, pady=10)
```
