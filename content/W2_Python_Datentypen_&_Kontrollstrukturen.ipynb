### 1. Erweiterte Datentypen in Python

Python nutzt die Funktion `type()`, um die Klasse eines Objekts zur Laufzeit zu bestimmen. Dies ist insbesondere bei der Arbeit mit dynamisch typisierten Variablen essenziell für das Debugging.

#### Sequenzielle Datentypen: Listen und Tupel

* **Listen (`list`):** Geordnete, veränderliche (mutable) Sammlungen. Sie erlauben den Zugriff über einen nullbasierten Index sowie das dynamische Hinzufügen von Elementen mittels `.append()`.
* **Tupel (`tuple`):** Geordnete, aber **unveränderliche** (immutable) Sammlungen. Sie werden häufig für Datensätze verwendet, deren Integrität während der Laufzeit geschützt werden muss (z.B. Konstanten oder Koordinaten).

```
# Listen-Operationen
primzahlen = [2, 3, 5, 7]
primzahlen.append(11)
primzahlen[0] = 1 # Mutation möglich

# Tupel-Operationen
koordinaten = (49.48, 8.46)
# koordinaten[0] = 50.0  # Erzeugt einen TypeError
```

#### Schlüssel-Wert-Speicher: Dictionaries (`dict`)

Ein Dictionary bildet eindeutige Schlüssel (Keys) auf beliebige Werte (Values) ab. Ab Python 3.7 ist die Einfügereihenfolge in Dictionaries garantiert.

```
studierender = {
    "name": "Mustermann",
    "matrikelnummer": 123456,
    "fakultaet": "BWL"
}
print(studierender["name"])
```

#### Zeichenketten und Formatierung

Strings (`str`) können mit f-Strings effizient formatiert werden, indem Ausdrücke innerhalb von geschweiften Klammern direkt in den String eingebettet werden.

```
name = "Student"
print(f"Willkommen im Modul 2, {name}.")
```

### 2. Bedingte Verzweigungen (`if-elif-else`)

Kontrollstrukturen ermöglichen es, Codeblöcke nur unter bestimmten Bedingungen auszuführen. In Python wird die Zugehörigkeit eines Codeblocks zu einer Bedingung ausschließlich durch **Einrückungen** (Indentation) definiert.

#### Struktur der Mehrfachverzweigung

Bedingungen werden sequentiell von oben nach unten geprüft. Sobald eine Bedingung `True` ergibt, wird der zugehörige Block ausgeführt und die restliche Struktur übersprungen.

```
# Beispiel: Validierung von Eingabewerten
testergebnis = 85

if testergebnis >= 90:
    bewertung = "Exzellent"
elif testergebnis >= 75:
    bewertung = "Gut"
elif testergebnis >= 50:
    bewertung = "Bestanden"
else:
    bewertung = "Nicht bestanden"

print(f"Ergebnis: {bewertung}")
```
