### 1. Die Programmiersprache Python

Python ist eine interpretierte High-Level-Sprache. Im Gegensatz zu kompilierten Sprachen (wie C++ oder Java) wird der Quellcode zur Laufzeit durch einen **Interpreter** Zeile für Zeile analysiert und ausgeführt.

**Eigenschaften des Interpreters:**

* **Plattformunabhängigkeit:** Der Code ist auf jedem System ausführbar, auf dem ein Interpreter installiert ist.
* **Interaktivität:** Über die Python-Shell können Befehle direkt getestet werden.
* **Fehlerdiagnose:** Laufzeitfehler werden unmittelbar an der betroffenen Code-Stelle gemeldet.

### 2. Variablen und Datentypen

In Python werden Variablen durch Zuweisung erstellt. Python nutzt **dynamische Typisierung**, was bedeutet, dass der Datentyp eines Objekts zur Laufzeit bestimmt wird.

#### Benennungskonventionen (PEP 8)

* Verwenden Sie `snake_case` (Kleinschreibung mit Unterstrichen).
* Namen müssen mit einem Buchstaben oder Unterstrich beginnen.
* Reservierte Schlüsselwörter (z.B. `if`, `while`, `pass`) dürfen nicht als Variablennamen verwendet werden.

#### Primäre Datentypen

| Typ | Bezeichnung | Beispiel |
| --- | --- | --- |
| `int` | Ganze Zahlen | `alter = 21` |
| `float` | Gleitkommazahlen | `preis = 19.99` |
| `str` | Zeichenketten | `name = "Mannheim"` |
| `bool` | Wahrheitswerte | `valid = True` |

```
# Beispiel: Variablenzuweisung und Typen
studierende_anzahl = 150          # int
durchschnittsnote = 1.7           # float
kurs_name = "Python Basiskurs"    # str
ist_aktiv = True                  # bool

print(type(studierende_anzahl))
```

### 3. Operatoren

#### Arithmetische Operatoren

Diese dienen der Durchführung mathematischer Berechnungen.

```
a = 10
b = 3

print(a / b)   # Division (ergibt float: 3.333...)
print(a // b)  # Ganzzahl-Division (3)
print(a % b)   # Modulo-Operator (Restwert: 1)
print(a ** b)  # Exponentiation (10^3 = 1000)
```

#### Vergleichsoperatoren

Vergleiche liefern stets einen booleschen Wert (`True` oder `False`) zurück.

* `==` Gleichheit
* `!=` Ungleichheit
* `>` / `<` Größer / Kleiner als
* `>=` / `<=` Größer / Kleiner gleich

#### Logische Operatoren

Zur Verknüpfung von Bedingungen.

* `and`: Wahr, wenn beide Bedingungen erfüllt sind.
* `or`: Wahr, wenn mindestens eine Bedingung erfüllt ist.
* `not`: Invertiert den Wahrheitswert.
