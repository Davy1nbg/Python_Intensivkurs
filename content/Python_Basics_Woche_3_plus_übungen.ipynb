## 1. Einführung: Was sind Schleifen und warum sind sie wichtig?

#### Definition und Bedeutung
Schleifen sind grundlegende Kontrollstrukturen in der Programmierung. Sie ermöglichen es, einen bestimmten Codeblock **wiederholt** auszuführen, solange eine bestimmte Bedingung erfüllt ist oder bis alle Elemente einer Sequenz durchlaufen wurden. Ohne Schleifen müssten wir Code oft redundant schreiben, was ihn länger, fehleranfälliger und schwerer wartbar machen würde.

#### Warum sind Schleifen so wichtig?
Stell dir vor, du müsstest 100 E-Mails versenden oder jeden Eintrag in einer langen Liste bearbeiten. Ohne Schleifen wäre das extrem mühsam! Schleifen automatisieren solche wiederkehrenden Aufgaben:

* **Datenverarbeitung:** Durchlaufen von Listen, Zeichenketten oder anderen Datensammlungen (z.B. alle Studenten in einer Liste begrüßen).
* **Berechnungen:** Wiederholte Anwendung von Formeln (z.B. Zinseszinsberechnung über mehrere Jahre).
* **Benutzerinteraktionen:** Warten auf gültige Benutzereingaben.
* **Simulationen und Spiele:** Aktualisieren von Zuständen in jedem Zeitschritt.

#### Arten von Schleifen in Python
Python bietet hauptsächlich zwei Arten von Schleifen:

* **`for`-Schleife:** Ideal, wenn man eine bekannte Anzahl von Wiederholungen hat oder über die Elemente einer Sequenz (z.B. Liste, String, `range()`) iterieren möchte.
* **`while`-Schleife:** Perfekt, wenn die Anzahl der Wiederholungen im Voraus unbekannt ist und die Schleife so lange laufen soll, wie eine bestimmte Bedingung wahr (`True`) ist.

## 2. Die `for`-Schleife: Iterieren mit System

Die `for`-Schleife wird verwendet, um über die Elemente einer **Sequenz** (wie eine Liste, ein Tupel, ein String) oder andere **iterierbare Objekte** (wie `range()`) zu gehen.

**Grundlegende Syntax:**

```
for element in sequenz:
    # Anweisungsblock, der für jedes Element ausgeführt wird
    print(element)
```

Hierbei nimmt die Variable `element` in jeder Wiederholung (Iteration) nacheinander den Wert eines jeden Elements aus der `sequenz` an.

#### Iterieren über Listen, Strings und Tupel

```
# Beispiel 1: Liste von Namen
namen = ["Anna", "Bernd", "Clara"]
for name in namen:
    print(f"Hallo, {name}!")

# Beispiel 2: Zeichen eines Strings
wort = "Python"
for buchstabe in wort:
    print(buchstabe)

# Beispiel 3: Elemente eines Tupels
koordinaten = (10, 20, 30)
for koordinate in koordinaten:
    print(f"Koordinate: {koordinate}")
```

#### Iterieren über Zahlenbereiche mit `range()`
Die `range()`-Funktion ist extrem nützlich, um Sequenzen von Zahlen zu erzeugen, über die man iterieren kann.

* `range(stop)`: Erzeugt Zahlen von 0 bis `stop-1`.

```
for i in range(5):  # Zahlen 0, 1, 2, 3, 4
        print(i)
```

* `range(start, stop)`: Erzeugt Zahlen von `start` bis `stop-1`.

```
for i in range(2, 6):  # Zahlen 2, 3, 4, 5
        print(i)
```

* `range(start, stop, step)`: Erzeugt Zahlen von `start` bis `stop-1`, mit einer Schrittweite von `step`.

```
for i in range(0, 11, 2):  # Zahlen 0, 2, 4, 6, 8, 10 (jeder zweite)
        print(i)
```

#### Iterieren über Dictionaries (kurzer Ausblick)
Man kann auch über Dictionaries iterieren:

```
benutzer = {"name": "Max", "alter": 30}
for schluessel in benutzer: # Iteriert standardmäßig über Schlüssel
    print(f"{schluessel}: {benutzer[schluessel]}")

# Oder direkt über Schlüssel-Wert-Paare:
for schluessel, wert in benutzer.items():
    print(f"{schluessel} -> {wert}")
```

## 3. Die `while`-Schleife: Wiederholen solange eine Bedingung gilt

Die `while`-Schleife führt einen Codeblock so lange aus, wie eine angegebene **Bedingung** wahr (`True`) ist.

**Grundlegende Syntax:**

```
while bedingung:
    # Anweisungsblock, der ausgeführt wird, solange bedingung True ist
    # Wichtig: Innerhalb der Schleife muss etwas geschehen,
    # das die bedingung irgendwann False werden lässt!
```

**Struktur einer typischen `while`-Schleife:**
1.  **Initialisierung:** Eine oder mehrere Variablen, die in der Bedingung verwendet werden, werden vor der Schleife initialisiert.
2.  **Bedingung:** Die Schleife läuft, solange diese Bedingung `True` ergibt.
3.  **Anweisungsblock:** Der Code, der wiederholt ausgeführt wird.
4.  **Aktualisierung:** Innerhalb des Blocks muss in der Regel eine Variable geändert werden, die Teil der Bedingung ist, um eine Endlosschleife zu vermeiden.

**Beispiele:**

```
# Beispiel 1: Zählen bis zu einem bestimmten Wert
count = 0
while count < 5:
    print(f"Aktueller Zähler: {count}")
    count += 1  # Aktualisierung: count erhöhen, um Endlosschleife zu vermeiden
print("Schleife beendet.")

# Beispiel 2: Benutzereingabe, bis ein Stoppwort eingegeben wird
eingabe = ""
while eingabe.lower() != "ende":
    eingabe = input("Gib etwas ein (oder 'ende' zum Beenden): ")
    if eingabe.lower() != "ende":
        print(f"Du hast eingegeben: {eingabe}")
print("Programm beendet.")
```

⚠️ **Achtung: Endlosschleifen!**
Wenn die Bedingung einer `while`-Schleife niemals `False` wird, entsteht eine **Endlosschleife**. Das Programm hängt dann fest. Achte immer darauf, dass die Bedingung beeinflusst wird und die Schleife ein definiertes Ende hat. (Notfalls Programm mit Strg+C im Terminal abbrechen).

## 4. Schleifensteuerung: `break`, `continue`, `pass`

Manchmal möchte man den normalen Ablauf einer Schleife beeinflussen.

#### `break`
Das Schlüsselwort `break` **beendet die aktuelle (innerste) Schleife sofort**, unabhängig davon, ob die Schleifenbedingung noch erfüllt ist. Die Programmausführung wird nach der Schleife fortgesetzt.

```
for i in range(10):  # Geplant von 0 bis 9
    if i == 5:
        print("Schleife wird bei i=5 abgebrochen.")
        break  # Verlässt die Schleife
    print(i)
# Ausgabe: 0, 1, 2, 3, 4, Schleife wird bei i=5 abgebrochen.
```

#### `continue`
Das Schlüsselwort `continue` **beendet den aktuellen Schleifendurchlauf sofort** und springt zum **Anfang des nächsten Durchlaufs**. Der restliche Code im aktuellen Durchlauf wird übersprungen.

```
for i in range(1, 8): # Zahlen 1 bis 7
    if i % 2 == 0:  # Wenn i gerade ist...
        print(f"Überspringe gerade Zahl: {i}")
        continue  # ...überspringe den print(i)-Befehl und mache mit nächstem i weiter
    print(f"Ungerade Zahl verarbeitet: {i}")
# Ausgabe: Ungerade Zahl verarbeitet: 1, Überspringe gerade Zahl: 2, Ungerade Zahl verarbeitet: 3, ...
```

#### `pass`
Das Schlüsselwort `pass` ist eine Nulloperation – es passiert nichts, wenn es ausgeführt wird. Es wird als Platzhalter verwendet, wo syntaktisch eine Anweisung erforderlich ist, man aber (noch) keine Logik implementieren möchte.

```
for i in range(5):
    if i == 3:
        # TODO: Hier später spezielle Logik für i=3 einfügen
        pass  # Ohne pass gäbe es hier einen Einrückungsfehler
    print(i)
```

#### Optional: Die `else`-Klausel in Schleifen
Python-Schleifen (`for` und `while`) können eine optionale `else`-Klausel haben. Der `else`-Block wird ausgeführt, **wenn die Schleife regulär beendet wurde** (d.h. bei `for` alle Elemente durchlaufen wurden oder bei `while` die Bedingung `False` wurde), **aber nicht, wenn die Schleife durch ein `break` vorzeitig verlassen wurde.**

```
for i in range(5):
    print(i)
    if i == 10: # Diese Bedingung wird nie wahr, kein break
        break
else:
    print("Schleife vollständig und ohne break durchlaufen.") # Wird ausgeführt

for i in range(5):
    print(i)
    if i == 2:
        break
else:
    print("Schleife vollständig durchlaufen.") # Wird NICHT ausgeführt
```

Dies kann nützlich sein, z.B. um zu signalisieren, ob eine Suche in einer Schleife erfolgreich war (und mit `break` beendet wurde) oder nicht.
