
# Block 3: Die erste Django "App" & ein simpler View (ca. 75-90 Min.) 🚀

**Ziel:** Die Studenten verstehen das Konzept einer Django-App, erstellen ihre erste App, schreiben eine einfache View-Funktion, die direkt HTML zurückgibt, und lernen, wie man diese View über eine URL erreichbar macht.


## 1. Was ist eine Django-App? (ca. 10-15 Minuten)

**Anknüpfung an Block 2:**
"Wir haben jetzt ein Django-Projekt (`mein_erstes_projekt`). Das ist wie das Fundament und das Dach unseres Hauses. Aber ein Haus ist leer ohne Zimmer, und jedes Zimmer hat einen bestimmten Zweck. In Django nennen wir diese 'Zimmer' oder funktionalen Einheiten **Apps**."

**Definition einer Django-App:**
* Eine Django-App ist ein **eigenständiges Modul** innerhalb eines Django-Projekts.
* Jede App ist dafür gedacht, eine **spezifische Funktionalität** der Webseite bereitzustellen.
* **Beispiele:**
    * In einer Nachrichten-Webseite könnte es eine App für Artikel, eine für Benutzerkommentare und eine für Umfragen geben.
    * In einem Online-Shop: eine App für Produkte, eine für den Warenkorb, eine für Kundenkonten.

**Vorteile von Apps:**
* **Modularität & Organisation:** Sie helfen, den Code übersichtlich zu halten. Man unterteilt ein großes Projekt in kleinere, logische und besser handhabbare Teile.
* **Wiederverwendbarkeit:** Eine gut geschriebene App (z.B. ein Umfrage-Tool) könnte potenziell auch in einem anderen Django-Projekt wiederverwendet werden.

**Analogien:**
* **Haus (Projekt) 🏠 → Zimmer (App) 🛋️:** Ein Haus (das Django-Projekt) hat verschiedene Zimmer: ein Wohnzimmer (vielleicht eine App für statische Seiten), eine Küche (eine App für Rezepte), ein Bad. Jedes Zimmer hat eine eigene Funktion, aber alle gehören zum selben Haus.
* **Lego-Set (Projekt) 🧱 → Spezialisierte Lego-Module (App) 🚀:** Man baut ein großes Raumschiff (Projekt) aus verschiedenen vorgefertigten oder selbst gebauten Modulen (Apps), z.B. ein Cockpit-Modul, ein Antriebs-Modul.

**Unsere erste App erstellen (Live-Coding):**

* **Wichtig:** Stellt sicher, dass eure virtuelle Umgebung (`venv`) aktiv ist und ihr euch im Hauptverzeichnis eures Django-Projekts befindet (dort, wo die Datei `manage.py` liegt).
* Der Befehl zum Erstellen einer neuen App lautet: `python manage.py startapp <appname>`
* Wir nennen unsere erste App `meine_app`:

    ```bash
    python manage.py startapp meine_app
    ```
* Django erstellt nun automatisch einen neuen Ordner namens `meine_app` mit einer vordefinierten Struktur.


**Die neue Ordnerstruktur von `meine_app/` kurz vorstellen:**

Wenn ihr jetzt in euer Projektverzeichnis schaut, seht ihr den neuen Ordner `meine_app/`. Werfen wir einen kurzen Blick hinein (nicht jedes Detail ist heute wichtig):
* `__init__.py`: Eine leere Datei, die Python signalisiert, dass dieser Ordner als Python-Paket behandelt werden soll.
* `admin.py`: Hier kann man Modelle für die Django-Admin-Oberfläche registrieren (für uns heute nicht relevant).
* `apps.py`: Enthält Konfigurationsinformationen für die App selbst.
* `migrations/`: Ein Ordner, der Skripte für Datenbankänderungen (Migrationen) enthält (für uns heute nicht relevant).
* `models.py`: Hier würde man Datenbankmodelle definieren, also die Struktur der Daten (für uns heute nicht relevant).
* `tests.py`: Hier kann man Tests für die App schreiben (für uns heute nicht relevant).
* `views.py`: **Diese Datei ist für uns heute sehr wichtig!** Hier schreiben wir die Python-Funktionen (oder Klassen), die auf Web-Anfragen reagieren und Antworten generieren.


**WICHTIG: Die neue App im Projekt bekannt machen (registrieren)!**

* Eine neu erstellte App muss dem Django-Projekt explizit mitgeteilt werden, dass sie existiert und genutzt werden soll.
* Öffnet dazu die Datei `mein_erstes_projekt/settings.py` (im Unterordner eures Projektnamens).
* Sucht die Liste namens `INSTALLED_APPS`.
* Fügt den Namen eurer App (`'meine_app'`) zu dieser Liste hinzu (am besten am Ende der von Django generierten Apps, aber vor eventuellen Drittanbieter-Apps):

````
# mein_erstes_projekt/settings.py

    INSTALLED_APPS = [
        'django.contrib.admin',
        'django.contrib.auth',
        'django.contrib.contenttypes',
        'django.contrib.sessions',
        'django.contrib.messages',
        'django.contrib.staticfiles',
        'meine_app',  # <-- Unsere neue App hier eintragen!
    ]
````

* Speichert die `settings.py`-Datei.

## 2. Der "Hello World" View (ca. 30-35 Minuten)

**Was ist ein View in Django? 🤔**
* Ein **View** (manchmal auch "Ansicht" genannt) ist das Herzstück der Logik in einer Django-App.
* Es ist typischerweise eine Python-Funktion (oder eine Klasse, aber wir starten mit Funktionen), die eine Web-Anfrage (`Request`) entgegennimmt und eine Web-Antwort (`Response`) zurückgibt.
* Hier wird entschieden, was passieren soll, wenn ein Nutzer eine bestimmte URL aufruft. Soll eine HTML-Seite angezeigt, Daten aus einer Datenbank geholt oder eine Berechnung durchgeführt werden?
* **Analogie:** Der **Koch** in unserer Restaurant-Analogie. Er erhält die Bestellung des Gastes (den Request), bereitet das gewünschte Gericht zu (verarbeitet die Anfrage, erzeugt vielleicht HTML) und gibt es dann zum Servieren frei (die Response).


**Unsere erste View-Funktion schreiben (Live-Coding):**

* Öffnet die Datei `meine_app/views.py`. Diese Datei ist anfangs meist leer oder enthält nur einen Kommentar.
* Wir schreiben jetzt eine sehr einfache Funktion, die eine "Hallo Welt"-Nachricht als HTML zurückgibt:

````
# meine_app/views.py

    from django.http import HttpResponse # Wichtig: HttpResponse muss importiert werden!

    def hallo_django_view(request):
        # Der Parameter 'request' ist ein Objekt, das Informationen über die
        # aktuelle Web-Anfrage vom Browser enthält (z.B. welcher Browser,
        # welche Daten mitgeschickt wurden etc.).
        # Für unsere einfache View nutzen wir es heute noch nicht aktiv.

        html_code = "<h1>Hallo Welt aus meiner Django App!</h1>"
        html_code += "<p>Das ist meine allererste, dynamisch mit Python erzeugte Webseite.</p>"
        # Kleiner Scherz, um request.META kurz zu zeigen (Zeit ist hier nicht wirklich dynamisch):
        client_ip = request.META.get('REMOTE_ADDR', 'Unbekannte IP')
        html_code += f"<p>Deine IP-Adresse ist: {client_ip}. Die aktuelle Zeit ist noch nicht dynamisch implementiert!</p>"

        # Wir erstellen eine HTTP-Antwort, die diesen HTML-Code enthält.
        return HttpResponse(html_code)
````

**Erklärungen zum Code:**
* `from django.http import HttpResponse`: Wir müssen die Klasse `HttpResponse` aus dem Modul `django.http` importieren. Ein Objekt dieser Klasse wird verwendet, um eine einfache textbasierte (oft HTML) Antwort an den Browser des Nutzers zu senden.
* `def hallo_django_view(request):`: Wir definieren eine Python-Funktion. Jede View-Funktion in Django muss als ersten Parameter ein `request`-Objekt akzeptieren. Dieses Objekt enthält alle Details der Anfrage, die vom Browser des Nutzers kam.
* `html_code = "..."`: Wir erstellen einen Python-String, der einfachen HTML-Code enthält. Hier seht ihr, wie Python direkt HTML generieren kann!
* `return HttpResponse(html_code)`: Unsere View-Funktion gibt ein `HttpResponse`-Objekt zurück. Dieses Objekt "verpackt" unseren `html_code`-String und Django sorgt dafür, dass er korrekt an den Browser gesendet wird.
* Speichert die `views.py`-Datei.


## 3. URL-Routing: Wie kommt der Aufruf zur Funktion? (ca. 25-30 Minuten)

**Das Problem:**
Wir haben jetzt eine schicke View-Funktion (`hallo_django_view`). Aber woher soll Django wissen, dass es genau diese Funktion aufrufen soll, wenn ein Nutzer eine bestimmte Adresse (URL) in seinen Browser eingibt, z.B. `http://127.0.0.1:8000/gruss/`?

**Die Lösung: URL-Routing (oder URLconf)**
* Django hat einen Mechanismus, um eingehende URLs (die Pfade nach der Domain, z.B. `/gruss/` oder `/artikel/42/`) bestimmten View-Funktionen zuzuordnen. Das passiert in speziellen Python-Dateien, die üblicherweise `urls.py` heißen.
* Man definiert Muster (Patterns) für URLs und welche View-Funktion bei einem passenden Muster ausgeführt werden soll.
* **Analogie:** Ein **Wegweiser-System** in einer großen Stadt oder die **Rezeption** in einem Hotel. Der Besucher (Browser mit URL) kommt an, und der Wegweiser/Rezeptionist (URL-Router) sagt ihm, in welchen Raum/zu welcher Abteilung (View-Funktion) er gehen muss.


**Schritt-für-Schritt zum URL-Routing (Live-Coding):**

**A. URLs für unsere App definieren (`meine_app/urls.py`):**
* Jede App kann (und sollte) ihre eigenen URL-Konfigurationen haben.
* Erstellt im Ordner eurer App (`meine_app/`) eine **neue Datei** namens `urls.py` (diese Datei wird von `startapp` nicht automatisch erstellt).
* Fügt folgenden Inhalt in `meine_app/urls.py` ein:

````
# meine_app/urls.py

    from django.urls import path
    from . import views  # Importiert die views.py aus dem aktuellen App-Verzeichnis ('meine_app')

    urlpatterns = [
        # Wenn jemand zur "Wurzel" dieser App-URL-Konfiguration kommt (was das genau ist,
        # definieren wir im nächsten Schritt im Projekt-urls.py),
        # dann rufe die Funktion 'hallo_django_view' aus unserer views.py auf.
        path('', views.hallo_django_view, name='hallo_seite'),
    ]
````

**Erklärungen zum Code:**
* `from django.urls import path`: Importiert die Funktion `path`, die wir zum Definieren eines URL-Musters verwenden.
* `from . import views`: Importiert das `views`-Modul (also unsere `views.py`-Datei) aus dem aktuellen Verzeichnis (dem `meine_app`-Ordner). Das `.` steht für das aktuelle Verzeichnis.
* `urlpatterns = [...]`: Eine Python-Liste, die alle URL-Muster für diese App enthält. Django durchsucht diese Liste.
* `path('', views.hallo_django_view, name='hallo_seite')`: Definiert ein einzelnes URL-Muster.
    * `''` (leerer String): Das eigentliche URL-Teilstück, auf das hier reagiert wird. Ein leerer String bedeutet hier "der Basispfad, unter dem diese App-URLs eingebunden werden". Wenn wir diese App-URLs z.B. unter `/gruss/` im Projekt einbinden, dann reagiert dieser Eintrag auf die genaue URL `/gruss/`.
    * `views.hallo_django_view`: Die View-Funktion, die aufgerufen werden soll, wenn dieses Muster passt.
    * `name='hallo_seite'`: Ein optionaler, aber sehr empfohlener Name für diese URL-Route. Mit diesem Namen kann man später in Templates oder anderem Python-Code dynamisch Links zu dieser URL erzeugen, ohne den Pfad hartcodieren zu müssen.


**B. App-URLs in die Projekt-URLs einbinden (`mein_erstes_projekt/urls.py`):**
* Jetzt müssen wir unserem Haupt-Projekt sagen, dass es die URL-Konfigurationen unserer `meine_app` berücksichtigen soll und unter welchem "Hauptpfad" diese erreichbar sein sollen.
* Öffnet die Datei `mein_erstes_projekt/urls.py` (die im Projektkonfigurationsordner, NICHT die, die ihr gerade erstellt habt).
* Passt den Code wie folgt an:

````
# mein_erstes_projekt/urls.py

    from django.contrib import admin
    from django.urls import path, include  # Wichtig: 'include' muss importiert werden!

    urlpatterns = [
        path('admin/', admin.site.urls),
        # Wenn die URL im Browser mit 'gruss/' beginnt,
        # dann leite die Anfrage und den Rest der URL
        # an die URL-Konfiguration in 'meine_app.urls' weiter.
        path('gruss/', include('meine_app.urls')),
    ]
````

**Erklärungen zum Code:**
* `from django.urls import path, include`: Stellt sicher, dass `include` importiert wird. Die `include()`-Funktion erlaubt es, URL-Konfigurationen aus anderen Apps einzubinden.
* `path('gruss/', include('meine_app.urls')):`
    * `'gruss/'`: Dies ist das URL-Präfix. Alle URLs, die im Browser mit `http://127.0.0.1:8000/gruss/` beginnen, werden von dieser Zeile "abgefangen".
    * `include('meine_app.urls')`: Sagt Django: "Für alles, was nach `gruss/` in der URL kommt (oder wenn es genau `gruss/` ist), ignoriere den `gruss/`-Teil und schaue für den Rest der URL-Verarbeitung bitte in der Datei `urls.py` der App `meine_app` nach weiteren Anweisungen."
* Speichert **beide** `urls.py`-Dateien.

**Testen! Die Spannung steigt! (Live-Coding)**

1.  **Entwicklungsserver starten (oder neu starten):**
    * Geht zurück in euer Terminal/Kommandozeile.
    * Wenn der Entwicklungsserver noch läuft, stoppt ihn mit `Strg + C` (bzw. `Ctrl + C`).
    * Startet ihn neu, damit er die Änderungen an den `urls.py`-Dateien erkennt:
        ```bash
        python manage.py runserver
        ```
    * Achtet auf Fehlermeldungen im Terminal. Wenn alles glattlief, sollte der Server ohne Probleme starten.

2.  **Im Browser aufrufen:**
    * Öffnet euren Webbrowser.
    * Gebt die Adresse ein, die wir gerade konfiguriert haben: `http://127.0.0.1:8000/gruss/`
    * Drückt Enter.

🎉 **ERFOLGSERLEBNIS!** 🎉 Ihr solltet jetzt eure "Hallo Welt aus meiner Django App!"-Nachricht im Browser sehen!


## 4. Kurze Erklärung: Was ist gerade passiert? Der Weg der Anfrage (ca. 5-10 Minuten)

Lasst uns den Weg der Anfrage gemeinsam nachverfolgen:

1.  **Dein Browser:** Du gibst `http://127.0.0.1:8000/gruss/` ein und drückst Enter.
2.  **HTTP Request:** Dein Browser schickt eine HTTP GET-Anfrage an den Django-Entwicklungsserver auf deinem Computer.
3.  **Django (Projekt `urls.py`):** Der Django-Server empfängt die Anfrage für den Pfad `/gruss/`.
    * Django schaut in die Datei `mein_erstes_projekt/urls.py`.
    * Es prüft die `urlpatterns`-Liste von oben nach unten.
    * Der Eintrag `path('admin/', ...)` passt nicht.
    * Der Eintrag `path('gruss/', include('meine_app.urls'))` passt! Der angefragte Pfad beginnt mit `gruss/`.
    * Django "schneidet" nun den passenden Teil (`gruss/`) vom angefragten Pfad ab. Übrig bleibt ein **leerer String** `''` (da die URL genau `http://.../gruss/` war).
    * Die `include()`-Funktion leitet die Anfrage (mit dem verbleibenden leeren Pfad `''`) an die URL-Konfiguration von `meine_app` weiter (also an `meine_app/urls.py`).
4.  **Django (App `urls.py`):** Django schaut nun in die Datei `meine_app/urls.py`.
    * Es prüft die `urlpatterns`-Liste dieser Datei.
    * Der Eintrag `path('', views.hallo_django_view, name='hallo_seite')` wird geprüft. Passt der verbleibende Pfad (der leere String `''`) auf das Muster `''`? Ja!
5.  **Django (Ausführung der View):** Django ruft jetzt die Funktion `hallo_django_view` auf, die in `meine_app/views.py` definiert ist. Es übergibt dabei automatisch das `request`-Objekt an die Funktion.
6.  **Deine View-Funktion (`hallo_django_view`):**
    * Deine Python-Funktion wird ausgeführt.
    * Sie erstellt den HTML-String `html_code`.
    * Sie gibt ein `HttpResponse`-Objekt zurück, das diesen `html_code` enthält.
7.  **HTTP Response:** Django nimmt dieses `HttpResponse`-Objekt und schickt es als Antwort zurück an deinen Browser.
8.  **Dein Browser:** Der Browser empfängt die HTTP-Antwort, die den HTML-Code enthält, und stellt diesen HTML-Code als sichtbare Webseite dar.


**Wichtige Konzepte dieses Blocks noch einmal hervorheben:**

* **App:** Ein Modul für eine bestimmte Funktionalität.
* **View:** Eine Python-Funktion, die einen Request bearbeitet und eine Response zurückgibt.
* **URL-Routing (`urls.py`):** Das System, das URLs bestimmten Views zuordnet.
* **`HttpResponse`:** Die einfachste Art, eine textbasierte (HTML) Antwort zu senden.

**Ausblick:**

Das war der erste Schritt, um dynamische Inhalte mit Django zu erzeugen. Von hier aus könnte man lernen, wie man Templates für komplexeres und schöneres HTML verwendet (möglicherweise in einem Block 4), wie man Daten aus Formularen empfängt oder mit Datenbanken interagiert.
