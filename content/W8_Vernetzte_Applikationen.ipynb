### 1. Persistente Datenhaltung mit SQLite

SQLite dient als lokales Archiv für Informationen, die über die Laufzeit des Programms hinaus erhalten bleiben müssen. Da SQLite keine Server-Infrastruktur benötigt, ist es die ideale Wahl für Desktop-Applikationen.

```
import sqlite3

def init_db():
    # Aufbau einer Verbindung zur lokalen Archiv-Datei
    conn = sqlite3.connect("wetter_archiv.db")
    cursor = conn.cursor()

    # Erstellung einer Historien-Tabelle für Temperaturdaten
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS history (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            ort TEXT,
            temperatur REAL,
            zeitpunkt TIMESTAMP DEFAULT CURRENT_TIMESTAMP
        )
    """)
    conn.commit()
    conn.close()

init_db()
```

### 2. HTTP mit `requests`: Das Internet als Datenquelle

Zur Kommunikation mit Web-Servern wird die Bibliothek `requests` eingesetzt. Sie ermöglicht es, Daten von APIs (Application Programming Interfaces) im JSON-Format zu "saugen" und in Python-Strukturen (Dictionaries) zu transformieren.

#### Beispiel: Abruf von Echtzeit-Wetterdaten

Wir nutzen **Open-Meteo**, eine API, die ohne Registrierung oder API-Key zugänglich ist.

```
import requests

def hole_temperatur(lat, lon):
    # API-Endpunkt für meteorologische Daten
    url = f"https://api.open-meteo.com/v1/forecast?latitude={lat}&longitude={lon}&current_weather=true"

    response = requests.get(url)
    if response.status_code == 200:
        daten = response.json()
        # Extraktion der Temperatur aus der verschachtelten JSON-Struktur
        return daten['current_weather']['temperature']
    return None

# Abruf für Koordinaten (Beispiel Mannheim: 49.48, 8.46)
temp = hole_temperatur(49.48, 8.46)
print(f"Aktuelle Temperatur in Mannheim: {temp}°C")
```
