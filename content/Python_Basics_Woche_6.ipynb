### 1. Einführung "Das Internet" & HTTP (ca. 45-60 Min.) 🌐

**Ziel:** Die Studenten verstehen das grundlegende Frage-Antwort-Spiel im Web und die Rolle von HTML als Struktursprache.

#### Was passiert, wenn ich eine Webseite aufrufe? (ca. 20-25 Minuten)

**Einstiegsfrage an die Studenten:**
"Was denkt ihr, passiert technisch genau in dem Moment, wenn ihr eine Webadresse wie `www.google.de` in euren Browser eingebt und Enter drückt, bis die Seite erscheint?"

* *Kurzes Brainstorming anregen, Stichworte sammeln (z.B. Internet, Kabel, Server, Google, Computer spricht mit anderem Computer...)*

#### Analogie: Der Restaurantbesuch 🍽️

Stellt euch vor, ihr wollt etwas essen gehen:
* **Ihr** (mit eurem Webbrowser) seid der **Kunde**.
* Das **Internet** ist wie das **Straßennetz oder das Telefonnetz**, das euch mit dem Restaurant verbindet.
* Der **Webserver** (z.B. ein Computer von Google) ist das **Restaurant mit dem Koch**.
* Die **Webseite** (z.B. die Google-Startseite) ist das **Gericht**, das ihr bestellt habt.

#### Die Fachbegriffe – Was steckt dahinter?

* **Client:**
    * Das seid ihr bzw. euer Webbrowser (Chrome, Firefox, Safari, Edge). Der Client ist derjenige, der eine **Anfrage stellt** (etwas haben will).
    * *Analogie: Der Kunde im Restaurant, der eine Bestellung aufgibt.*

* **Server:**
    * Ein (oft sehr leistungsstarker) Computer, der irgendwo auf der Welt steht und darauf wartet, Anfragen von Clients zu bekommen. Auf ihm ist die Webseite gespeichert. Der Server liefert eine **Antwort**.
    * *Analogie: Das Restaurant bzw. der Koch, der die Bestellung entgegennimmt, das Gericht zubereitet und serviert.*

* **IP-Adresse & DNS** (kurz erwähnen, nicht vertiefen!):
    * Computer im Internet finden sich nicht über Namen wie `www.google.de`, sondern über eindeutige Nummern, sogenannte **IP-Adressen** (z.B. `172.217.160.78`).
    * Das **DNS (Domain Name System)** ist wie ein riesiges Telefonbuch für das Internet. Es übersetzt den Namen (`www.google.de`) in die zugehörige IP-Adresse.
    * *Analogie: Ihr kennt den Namen des Restaurants, aber um es im Navi einzugeben oder anzurufen, braucht ihr die genaue Adresse oder Telefonnummer. Das DNS liefert diese.*

* **Request (Anfrage):**
    * Euer Browser (Client) schickt eine Anfrage an den Server, auf dem die gewünschte Webseite liegt. Z.B.: "Lieber Google-Server, ich hätte gerne die Startseite."
    * *Analogie: Eure Bestellung beim Kellner: "Ich hätte gerne die Pizza Salami."*

* **Response (Antwort):**
    * Der Server verarbeitet die Anfrage und schickt eine Antwort zurück. Im Idealfall ist das die angeforderte Webseite. Es kann aber auch eine Fehlermeldung sein ("Seite nicht gefunden").
    * *Analogie: Der Kellner bringt euch die Pizza Salami oder sagt: "Tut mir leid, Pizza Salami ist heute aus."*

#### HTTP/HTTPS – Die Sprache des Webs 🗣️

* Damit Client und Server sich verstehen, brauchen sie eine gemeinsame Sprache. Diese Sprache heißt **HTTP (HyperText Transfer Protocol)**.
* **HTTPS** ist die sichere, verschlüsselte Variante davon (das "S" steht für Secure). Das ist heute der Standard.
* Die häufigste Art von HTTP-Anfrage ist eine **GET-Anfrage**: "Ich möchte gerne diese Ressource (z.B. Webseite, Bild) bekommen."
* *Analogie: Die Sprache, in der Kunde und Kellner miteinander sprechen (z.B. Deutsch). Die GET-Anfrage ist wie "Ich möchte die Speisekarte" oder "Ich bekomme..."*

Stellt euch das so vor:

```
Euer Computer (Client / Browser)        <---- Das Internet ---->        Webserver (z.B. von Google)
       |                                                                      ^
       | --- Request (z.B. GET / HTTP) -->                                    |
       |                                                                      |
       | <-- Response (z.B. HTML-Seite / HTTP) ---                            |
       v                                                                      |
Zeigt Webseite an                                                       Liefert Webseite
```

### 2. Was ist HTML? Die Struktur einer Webseite (ca. 25-35 Minuten) 📜

**Überleitung:**
"Okay, wir wissen jetzt, dass der Client eine Webseite vom Server anfordert und eine Antwort bekommt. Aber was genau schickt der Server da eigentlich zurück? In den meisten Fällen ist das ein Dokument, das in **HTML** geschrieben ist."

#### HTML – Die Sprache der Webseiten-Struktur

* **H**yper**T**ext **M**arkup **L**anguage
    * **HyperText:** Text, der nicht nur linear gelesen wird, sondern auch **Links** zu anderen Texten (oder Webseiten) enthalten kann. Das "Hyper" steht für diese Verknüpfungen.
    * **Markup Language (Auszeichnungssprache):** Eine Sprache, die Text mit speziellen Markierungen (sogenannten **Tags**) versieht, um ihm **Struktur und Bedeutung** zu geben.

* **Wichtig:** HTML ist **keine Programmiersprache** wie Python!
    * Es hat keine Variablen, Schleifen oder komplexe Logik.
    * Es beschreibt, **was etwas ist** (z.B. eine Überschrift, ein Absatz, ein Bild).

#### Analogie: Das Gerüst eines Hauses oder der Bauplan 🏗️

* **HTML-Tags** sind wie die Bauelemente eines Hauses: Es gibt Fundament (`<html>`), Wände, Stockwerke (`<body>`), Räume, Türen (`<a>`), Fenster (`<img>`).
* HTML sagt dem Browser (dem Bauarbeiter), wo die Überschrift hinkommt, wo ein Textabsatz beginnt und endet, wo ein Bild eingefügt wird oder wo ein Link zu einer anderen Seite ist.
* Es geht primär um die **Struktur und den Inhalt**, noch nicht um das Aussehen (Farben, Schriftarten – das macht später CSS).

#### Beispiel – Wir schauen unter die Haube einer Webseite:

Öffnet eine sehr einfache Webseite im Browser. Ihr könnt dazu den folgenden HTML-Code in eine Datei namens `beispiel.html` speichern und diese dann im Browser öffnen:

```html
<!DOCTYPE html>
<html>
<head>
    <title>Meine erste Webseite</title>
</head>
<body>
    <h1>Hallo Welt aus HTML!</h1>
    <p>Dies ist ein einfacher Textabsatz.</p>
    <p>Man kann auch <a href="https://www.python.org">Links</a> einbauen oder Bilder.</p>
</body>
</html>
```

**So schaut ihr euch den Quelltext an:**
* Macht einen Rechtsklick auf die geöffnete Webseite im Browser.
* Wählt eine Option wie "Seitenquelltext anzeigen", "View Page Source" oder ähnlich (variiert je nach Browser).

#### Die wichtigsten HTML-Tags im Überblick:

Geht die Tags gemeinsam durch und erklärt ihre Bedeutung (Prinzip verstehen, nicht auswendig lernen!):

* `<!DOCTYPE html>`: Deklaration, die dem Browser sagt: "Achtung, es folgt HTML5-Code."
* `<html> ... </html>`: Das Wurzelelement. Alles andere ist hier drin.
* `<head> ... </head>`: Der "Kopf" der Seite. Enthält Meta-Informationen, die nicht direkt im Browserfenster sichtbar sind (z.B. der Titel für den Browser-Tab, Angaben für Suchmaschinen).
    * `<title> ... </title>`: Der Text, der oben im Browser-Tab oder Fenster-Titel angezeigt wird.
* `<body> ... </body>`: Der "Körper" der Seite. Hier steht alles, was im Browserfenster sichtbar ist.
* `<h1> ... </h1>` (bis `<h6>`): Überschriften. `<h1>` ist die wichtigste/größte Überschrift.
* `<p> ... </p>`: Ein Textabsatz (Paragraph).
* `<a> ... </a>`: Ein Link (Anchor). Das Attribut `href="URL"` gibt an, wohin der Link führt.
    * Beispiel: `<a href="https://www.python.org">Besuche Python.org</a>`
* `<img>`: Fügt ein Bild ein.
    * `src="PFAD_ZUM_BILD"` gibt die Quelle des Bildes an.
    * `alt="BESCHREIBUNG"` ist ein Alternativtext (wichtig für Barrierefreiheit und falls das Bild nicht lädt).
    * Beispiel: `<img src="logo.png" alt="Firmenlogo">` (benötigt eine Bilddatei `logo.png` im selben Verzeichnis)


**Kernbotschaft zu HTML:**
* HTML gibt Webseiten ihre **Grundstruktur** und definiert die **Bedeutung (Semantik)** der Inhalte.
* Browser lesen diesen HTML-Code und stellen ihn **grafisch dar**.

"Wer möchte, kann jetzt schnell einen Texteditor öffnen (Notepad, VS Code, was auch immer ihr habt), den HTML-Code von eben abtippen oder kopieren, als `meineSeite.html` speichern und diese Datei dann im Browser öffnen. Ihr werdet sehen: Der Browser macht daraus eine Webseite!"


#### Ausblick – Was kommt noch dazu?
*(Nur je 1 Satz pro Punkt, um Neugier zu wecken)*

* **CSS (Cascading Style Sheets):** Nachdem HTML das Gerüst gebaut hat, kommt CSS und macht alles hübsch (Farben, Schriftarten, Layout).
    * *"HTML ist das Skelett, CSS sind die Kleider und das Make-up."* 💄👗
* **JavaScript:** Wenn die Seite nicht nur gut aussehen, sondern auch interaktiv sein soll (z.B. auf Klicks reagieren, Animationen zeigen), dann kommt JavaScript ins Spiel.
    * *"HTML ist das Skelett, CSS die Kleider, JavaScript das Gehirn und die Muskeln."* 🧠💪
* **Django & Python:** Und wo kommt jetzt Django ins Spiel? Django hilft uns **serverseitig** mit Python-Logik, dieses HTML (und später auch CSS und JavaScript-Referenzen) **dynamisch zu erzeugen** und an den Browser zu schicken! サーバー側 ✨
