### 1. Einführung in die Iterationskonzepte

Iterationen ermöglichen die wiederholte Ausführung eines Codeblocks. Dies ist essenziell für die Automatisierung repetitiver Aufgaben und die effiziente Verarbeitung von Datensammlungen.

**Klassifizierung der Schleifentypen in Python:**

* **Zähl- oder Sequenzschleifen (`for`):** Werden eingesetzt, wenn die Anzahl der Durchläufe bekannt ist oder über eine endliche Menge von Elementen (z.B. eine Liste) iteriert wird.
* **Bedingungsschleifen (`while`):** Finden Anwendung, wenn die Anzahl der Wiederholungen von einer dynamischen Bedingung abhängt, die erst zur Laufzeit evaluiert wird.

---

### 2. Die `for`-Schleife

Die `for`-Schleife in Python ist ein mächtiges Werkzeug zur Iteration über **iterierbare Objekte** (Iterables).

#### Iteration über Kollektionen

In jeder Iteration nimmt die Laufvariable automatisch den Wert des nächsten Elements der Sequenz an.

```
# Iteration über eine Liste
fakultaeten = ["VWL", "BWL", "Informatik"]
for fach in fakultaeten:
    print(f"Fachbereich: {fach}")

# Iteration über ein Dictionary
studierende = {"ID01": "Anna", "ID02": "Bernd"}
for key, value in studierende.items():
    print(f"Matrikelnummer: {key}, Name: {value}")
```

#### Erzeugung von Zahlenbereichen mit `range()`

Die Funktion `range()` generiert eine Sequenz von Ganzzahlen.

* `range(stop)`: 0 bis .
* `range(start, stop)`:  bis .
* `range(start, stop, step)`:  bis  mit definierter Schrittweite.

```
# Beispiel: Jeder zweite Wert von 0 bis 8
for i in range(0, 10, 2):
    print(i)
```

### 3. Die `while`-Schleife

Die `while`-Schleife wird ausgeführt, solange eine logische Bedingung den Wahrheitswert `True` liefert.

**Strukturelle Anforderungen:**

1. **Initialisierung:** Definition der Ausgangslage vor Schleifeneintritt.
2. **Abbruchbedingung:** Logischer Ausdruck, der die Fortführung prüft.
3. **Inkrementierung/Aktualisierung:** Modifikation der Kontrollvariable innerhalb der Schleife zur Vermeidung von Endlosschleifen.

```
grenzwert = 3
zaehler = 0

while zaehler < grenzwert:
    print(f"Durchlauf Nummer: {zaehler}")
    zaehler += 1
```

---

### 4. Steuerungselemente: `break`, `continue` und `pass`

Zur Feinsteuerung innerhalb von Iterationen bietet Python spezifische Schlüsselwörter:

* **`break`:** Bricht die Schleife unmittelbar ab und setzt die Programmausführung nach dem Schleifenblock fort.
* **`continue`:** Beendet lediglich den aktuellen Durchlauf und springt sofort zur Prüfung der nächsten Iterationsbedingung.
* **`pass`:** Dient als syntaktischer Platzhalter für noch zu implementierende Logik, ohne die Ausführung zu beeinflussen.
