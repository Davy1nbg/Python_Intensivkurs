### 1. Einführung in Python

#### Was ist Python?
Python ist eine **High-Level-Programmiersprache**, die für ihre **Einfachheit**, **Lesbarkeit** und **Vielseitigkeit** bekannt ist. Sie wurde Anfang der 1990er Jahre von Guido van Rossum entwickelt. "High-Level" bedeutet, dass die Sprache näher an der menschlichen Sprache ist als an der Maschinensprache des Computers, was das Programmieren erleichtert. Python wird in sehr vielen Bereichen eingesetzt:
* **Webentwicklung:** Mit Frameworks wie Django oder Flask.
* **Datenanalyse und Wissenschaft:** Mit Bibliotheken wie NumPy, Pandas, SciPy.
* **Künstliche Intelligenz (KI) und Maschinelles Lernen (ML):** Mit Bibliotheken wie TensorFlow, PyTorch, scikit-learn.
* **Automatisierung von Aufgaben:** Skripte für Systemadministration, Dateiverarbeitung etc.
* **Spieleentwicklung:** Einfache Spiele oder Prototypen.
* **Desktop-Anwendungen:** Mit Bibliotheken wie Tkinter oder PyQt.

#### Interpreter vs. Kompilierte Sprachen
Python ist eine **interpretierte Sprache**. Das bedeutet, dass der Python-Code nicht vor der Ausführung komplett in Maschinencode übersetzt (kompiliert) wird, sondern ein spezielles Programm, der **Python-Interpreter**, den Code **Zeile für Zeile liest und direkt ausführt**.
* **Vorteile des Interpreters:**
    * Schneller Entwicklungszyklus: Code kann sofort ausgeführt und getestet werden.
    * Einfache Fehlerbehebung: Fehler werden oft direkt an der Stelle angezeigt, wo sie auftreten.
    * Plattformunabhängigkeit: Der gleiche Python-Code kann auf verschiedenen Betriebssystemen ausgeführt werden, solange ein passender Interpreter installiert ist.
* **Unterschied zu kompilierten Sprachen (z.B. C++, Java):** Bei kompilierten Sprachen muss der gesamte Code erst durch einen Compiler in eine ausführbare Datei umgewandelt werden, bevor er ausgeführt werden kann.

### 2. Installation von Python

Bevor wir mit dem Programmieren beginnen können, musst du Python auf deinem Computer installieren. Die Installation ist in der Regel unkompliziert.

#### Für Windows:
1.  Besuche die offizielle Python-Website: [https://www.python.org/downloads/](https://www.python.org/downloads/)
2.  Lade den neuesten Installer für Windows herunter (achte auf 32-bit oder 64-bit, passend zu deinem System).
3.  Führe die heruntergeladene `.exe`-Datei aus.
4.  **WICHTIG:** Kreuze im ersten Fenster des Installers die Option **"Add Python to PATH"** an. Dies ist entscheidend, damit du Python-Befehle (wie `python` oder `pip`) von jedem beliebigen Ort in der Kommandozeile (Terminal/CMD) ausführen kannst, ohne jedes Mal den vollen Installationspfad angeben zu müssen. Es erleichtert den Start erheblich.
5.  Klicke auf "Install Now" und folge den Anweisungen.

#### Für macOS:
1.  Besuche die offizielle Python-Website: [https://www.python.org/downloads/](https://www.python.org/downloads/)
2.  Lade den neuesten macOS Installer herunter.
3.  Öffne die heruntergeladene `.pkg`-Datei und folge den Installationsanweisungen. Python 3 ist auf vielen macOS-Systemen bereits vorinstalliert. Es ist jedoch empfehlenswert, die neueste Version von der offiziellen Webseite zu installieren, um von aktuellen Funktionen, Fehlerbehebungen und Sicherheitsupdates zu profitieren.

#### Für Linux:
Python 3 ist auf den meisten Linux-Distributionen standardmäßig vorinstalliert. Du kannst die Installation über den Paketmanager deiner Distribution überprüfen oder eine neuere Version installieren:
* **Debian/Ubuntu:** Öffne ein Terminal und gib ein: `sudo apt update && sudo apt install python3 python3-pip`
* **Fedora:** Öffne ein Terminal und gib ein: `sudo dnf install python3 python3-pip`
* **Arch Linux:** Öffne ein Terminal und gib ein: `sudo pacman -S python python-pip`

#### Verifizierung der Installation:
Öffne nach der Installation ein Terminal oder die Eingabeaufforderung (Windows: Suche nach "cmd" oder "PowerShell"). Gib einen der folgenden Befehle ein und drücke Enter:
```bash
python --version
# oder falls das nicht funktioniert:
python3 --version
```
Es sollte die installierte Python-Version angezeigt werden (z.B. Python 3.10.4). Wenn ein Fehler auftritt, überprüfe, ob du "Add Python to PATH" (Windows) ausgewählt hast oder ob die Installation erfolgreich war.

### 3. Erste Schritte mit dem Python-Interpreter (Shell)

```plaintext

```
**Interaktive Ausführung:**

### 4. Ausführen von Python-Skripten
Für komplexere Programme schreibst du deinen Code in eine Datei (ein sogenanntes Skript) und führst diese Datei dann mit dem Interpreter aus.

1.  Öffne einen einfachen Texteditor (wie Notepad, VS Code, Sublime Text, etc.).
2.  Schreibe deinen Python-Code in die Datei.
3.  Speichere die Datei mit der Endung `.py` (z.B. `mein_erstes_skript.py`). Achte darauf, wo du die Datei speicherst.
4.  Öffne dein Terminal/Eingabeaufforderung.
5.  Navigiere zu dem Verzeichnis, in dem du die Datei gespeichert hast (mit dem Befehl `cd <Verzeichnisname>`). Zum Beispiel, wenn dein Skript im Ordner `PythonProjekte` auf deinem Desktop liegt, könnte der Befehl so aussehen: `cd Desktop/PythonProjekte` (macOS/Linux) oder `cd Desktop\PythonProjekte` (Windows). Die genaue Navigation hängt von deiner Ordnerstruktur ab.
6.  Führe das Skript mit dem Python-Interpreter aus:
    ```bash
    python mein_erstes_skript.py
    # oder
    python3 mein_erstes_skript.py
    ```
**Beispiel: `hallo_welt.py`**

````python
# Dies ist ein Kommentar - Python ignoriert diese Zeile
print("Hallo, Uni Mannheim!")
print("Das ist mein erstes Python-Skript.")
````

Wenn du dieses Skript ausführst, wird die Ausgabe im Terminal angezeigt.

### 5. Paketverwaltung mit pip
Python hat eine riesige Gemeinschaft, die nützliche Bibliotheken oder Pakete entwickelt. Das sind Sammlungen von Code, die andere geschrieben haben, um bestimmte Aufgaben zu erleichtern (z.B. für Webanfragen, Datenanalyse, etc.).

`pip` ist der Standard-Paketmanager für Python. Er wird verwendet, um diese externen Pakete herunterzuladen und zu installieren. `pip` wird normalerweise zusammen mit Python installiert.

**Wichtige `pip`-Befehle im Terminal:**
* **Paket installieren:**
    ```bash
    pip install <paketname>
    # Beispiel: pip install requests  # Installiert die requests-Bibliothek für Webanfragen
    ```
* **Installierte Pakete anzeigen:**
    ```bash
    pip list
    ```
* **Paket deinstallieren:**
    ```bash
    pip uninstall <paketname>
    ```
* **Paket aktualisieren:**
    ```bash
    pip install --upgrade <paketname>
    ```
Du wirst `pip` später im Kurs verwenden, wenn wir externe Bibliotheken nutzen.

### 6. Variablen
**Was sind Variablen?**
Stell dir Variablen wie beschriftete Boxen im Speicher deines Computers vor. Du kannst Daten (Werte) in diese Boxen legen und ihnen einen Namen (den Variablennamen) geben, um später auf die Daten zuzugreifen oder sie zu ändern.

In Python musst du Variablen nicht explizit deklarieren oder ihren Datentyp vorher festlegen. Du erstellst eine Variable einfach, indem du ihr zum ersten Mal einen Wert zuweist. Python erkennt den Datentyp automatisch (dynamische Typisierung).

**Zuweisung von Werten**
Mit dem Gleichheitszeichen (`=`) weist du einer Variable einen Wert zu.

````python
# Eine Variable namens 'alter' erstellen und ihr den Wert 20 zuweisen (Integer)
alter = 20  # alter ist vom Typ Integer

# Eine Variable namens 'name' erstellen und ihr den Wert "Anna" zuweisen (String)
name = "Anna"  # name ist vom Typ String

# Eine Variable namens 'groesse' erstellen und ihr den Wert 1.75 zuweisen (Float)
groesse = 1.75  # groesse ist vom Typ Float

# Eine Variable namens 'ist_student' erstellen und ihr den Wert True zuweisen (Boolean)
ist_student = True  # ist_student ist vom Typ Boolean

# Du kannst den Wert einer Variable jederzeit ändern
alter = alter + 1  # alter ist jetzt 21
name = "Hannah"  # name ist jetzt "Hannah"

print(alter)
print(name)
````

**Benennungskonventionen für Variablen**
Gute Variablennamen machen deinen Code lesbarer.

* **Snake_case:** Die empfohlene Konvention in Python. Variablennamen werden in Kleinbuchstaben geschrieben, Wörter werden durch Unterstriche getrennt (z.B. `anzahl_studenten`, `gesamtpreis`).
* **Aussagekräftige Namen:** Wähle Namen, die klar beschreiben, wofür die Variable verwendet wird (z.B. `benutzername` statt `bn`, `temperatur_celsius` statt `t`).
* **Regeln:**
    * Variablennamen dürfen nur Buchstaben (a-z, A-Z), Ziffern (0-9) und Unterstriche (`_`) enthalten.
    * Sie dürfen nicht mit einer Ziffer beginnen.
    * Sie dürfen keine Python-Schlüsselwörter (reservierte Wörter wie `if`, `for`, `while` etc.) sein und sollten nicht mit Namen von wichtigen eingebauten Funktionen (wie `print`) kollidieren. Eine Liste der Schlüsselwörter findest du in der Python-Dokumentation oder wirst sie im Laufe des Kurses kennenlernen.

**Grundlegende Datentypen (hier kurz vorgestellt und in Modul 2 vertieft):**
Obwohl Python den Typ automatisch erkennt, ist es wichtig, die grundlegenden Datentypen zu kennen:

* **Integer (`int`):** Ganze Zahlen (z.B. `10`, `-5`, `0`).
* **Float (`float`):** Gleitkommazahlen (Zahlen mit Dezimalpunkt) (z.B. `3.14`, `-0.5`, `2.0`).
* **String (`str`):** Text, der in Anführungszeichen steht (einfach `'...'` oder doppelt `"..."`) (z.B. `"Hallo"`, `'Python Kurs'`).
* **Boolean (`bool`):** Wahrheitswerte, entweder `True` oder `False`.

### 7. Operatoren
Operatoren sind Symbole, die Operationen auf Werten und Variablen durchführen.

**Arithmetische Operatoren**
Werden für mathematische Berechnungen verwendet.

````python
a = 10
b = 3

print(a + b)   # Addition: 13
print(a - b)   # Subtraktion: 7
print(a * b)   # Multiplikation: 30
print(a / b)   # Division (ergibt immer ein float): 3.333...
print(a // b)  # Ganzzahl-Division (ergibt den ganzzahligen Teil): 3
print(a % b)   # Modulo (Rest der Division): 1 (weil 10 = 3*3 + 1)
print(a ** b)  # Potenz (a hoch b): 1000 (10*10*10)
````

**Zuweisungsoperatoren**
Diese Operatoren kombinieren eine Operation mit einer Zuweisung. Sie sind eine Kurzform.

````python
x = 5
print(x)  # 5

x += 3  # Das ist das Gleiche wie: x = x + 3
print(x)  # 8

x -= 2  # Das ist das Gleiche wie: x = x - 2
print(x)  # 6

x *= 4  # Das ist das Gleiche wie: x = x * 4
print(x)  # 24

x /= 3  # Das ist das Gleiche wie: x = x / 3
print(x)  # 8.0 (Division ergibt immer float)

x //= 2  # Das ist das Gleiche wie: x = x // 2
print(x)  # 4.0 (Ergebnis bleibt float nach vorheriger Division)

y = 10
y %= 3  # Das ist das Gleiche wie: y = y % 3
print(y)  # 1

z = 2
z **= 3  # Das ist das Gleiche wie: z = z ** 3
print(z)  # 8
````

**Vergleichsoperatoren**
Werden verwendet, um Werte zu vergleichen. Sie geben immer einen Booleschen Wert zurück (`True` oder `False`).

````python
a = 10
b = 3
c = 10

print(a == b)  # Gleichheit: False (10 ist nicht gleich 3)
print(a == c)  # Gleichheit: True (10 ist gleich 10)
print(a != b)  # Ungleichheit: True (10 ist ungleich 3)
print(a > b)   # Größer als: True (10 ist größer als 3)
print(a < b)   # Kleiner als: False (10 ist nicht kleiner als 3)
print(a >= c)  # Größer oder gleich: True (10 ist größer oder gleich 10)
print(b <= a)  # Kleiner oder gleich: True (3 ist kleiner oder gleich 10)



a = True (1 = Spannung)
b = False (0 = Transistor hat keine Spannung)
c = False

var_1 = (a and b) or (a < b or a > c)
var_1 = False or (a < b or a > c)
var_1 = False or False or True
var_1 = False or True
var_1 = True

print(var_1) 

````

**Logische Operatoren**
Werden verwendet, um Boolesche Werte oder Ausdrücke zu kombinieren.

````python
hat_geld = True
ist_sonnig = False

print(hat_geld and ist_sonnig)  # Logisches UND: True nur, wenn BEIDE True sind -> False
print(hat_geld or ist_sonnig)   # Logisches ODER: True, wenn MINDESTENS EINER True ist -> True
print(not hat_geld)             # Logisches NICHT: Kehrt den Wert um -> False

# Beispiel mit Vergleichen:
alter = 25
print(alter > 18 and alter < 30)  # True (25 ist größer als 18 UND kleiner als 30)
print(alter < 18 or alter > 65)   # False (25 ist weder kleiner als 18 noch größer als 65)
````

### 8. Entwicklungsumgebungen (IDEs) und Tools
Während du Python-Skripte mit einem einfachen Texteditor und dem Terminal ausführen kannst, bieten Integrierte Entwicklungsumgebungen (IDEs) und erweiterte Editoren viele Vorteile, die das Programmieren erleichtern:

* **Syntax-Hervorhebung:** Code wird farblich unterschiedlich dargestellt, was die Lesbarkeit verbessert.
* **Code-Vervollständigung:** Die IDE schlägt Code vor, während du tippst.
* **Fehlererkennung:** Tippfehler oder Syntaxfehler werden oft schon während des Schreibens markiert.
* **Debugging:** Werkzeuge, um Fehler im Code Schritt für Schritt zu finden und zu beheben.
* Integrierte Terminals und Ausführungsmöglichkeiten.

Hier sind einige gängige Optionen:

**IDLE (Integrated Development and Learning Environment)**
* **Was es ist:** IDLE ist die Standard-IDE, die mit der Python-Installation mitgeliefert wird. Sie ist einfach und gut für Anfänger geeignet.
* **Funktionen:** Bietet einen interaktiven Shell-Modus und einen Editor zum Schreiben und Ausführen von Skripten.
* **Starten:** Nach der Installation findest du IDLE im Startmenü (Windows) oder in den Anwendungen (macOS).

**VS Code (Visual Studio Code)**
* **Was es ist:** Ein sehr beliebter, kostenloser und vielseitiger Code-Editor von Microsoft.
* **Vorteile:** Sehr anpassbar durch Erweiterungen, unterstützt viele Programmiersprachen. Für Python musst du die offizielle Python-Erweiterung installieren.
* **Link:** [https://code.visualstudio.com/](https://code.visualstudio.com/)

**PyCharm**
* **Was es ist:** Eine leistungsstarke, speziell für Python entwickelte IDE von JetBrains.
* **Vorteile:** Bietet sehr fortgeschrittene Funktionen für professionelle Entwicklung (Code-Analyse, Debugging, Integration mit Frameworks). Es gibt eine kostenlose Community-Edition, die für den Anfang ausreicht.
* **Link:** [https://www.jetbrains.com/pycharm/](https://www.jetbrains.com/pycharm/)

**Jupyter Notebooks**
* **Was es ist:** Eine webbasierte interaktive Umgebung, ideal für Datenanalyse, wissenschaftliches Rechnen und das Teilen von Code mit Erklärungen.
* **Vorteile:** Ermöglicht die Kombination von Code, Text (Markdown), Gleichungen und Visualisierungen in einem einzigen Dokument. Code kann in einzelnen "Zellen" ausgeführt werden.
* **Link:** [https://jupyter.org/](https://jupyter.org/) (Oft installiert über Anaconda Distribution oder mit `pip install notebook`)

Für den Anfang kannst du gut mit IDLE starten, da es bereits installiert ist. Später wirst du vielleicht die Vorteile von VS Code oder PyCharm schätzen lernen.
