### 1. Datentypen in Python

Python bietet eine Vielzahl von eingebauten Datentypen. Jeder Datentyp bestimmt, welche Art von Werten eine Variable speichern kann und welche Operationen darauf ausgeführt werden können. Mit der eingebauten Funktion `type(eine_variable)` könnt ihr jederzeit den Datentyp einer Variable im Code überprüfen, was besonders beim Lernen und Debuggen hilfreich ist.

Die wichtigsten Datentypen für den Anfang sind:

#### Zahlen (`int`, `float`) und Boolesche Werte (`bool`)

* **Integer (`int`):** Stellt ganze Zahlen ohne Nachkommastellen dar.
    * Beispiele: `alter = 35`, `anzahl_artikel = 10`, `temperatur = -5`
    *

```
x = 10
print(type(x))  # Ausgabe: <class 'int'>
```

* **Float (`float`):** Stellt Gleitkommazahlen dar, also Zahlen mit Nachkommastellen (Dezimaltrennzeichen ist der Punkt `.`).
    * Beispiele: `pi = 3.14159`, `preis = 19.99`, `koerpergroesse = 1.78`
    *

```
y = 3.14
print(type(y))  # Ausgabe: <class 'float'>
```

* **Boolean (`bool`):** Repräsentiert die Wahrheitswerte `True` (wahr) oder `False` (falsch). Boolesche Werte sind das Ergebnis von Vergleichen (siehe Modul 1) und die Grundlage für Entscheidungen in Verzweigungen.
    * Beispiele: `ist_volljaehrig = True`, `hat_bestanden = False`, `licht_an = (raum_helligkeit < 50)`
    *

```
ist_aktiv = True
print(type(ist_aktiv))  # Ausgabe: <class 'bool'>
```

* **Complex (`complex`):** (Kurz zur Kenntnisnahme) Python unterstützt auch komplexe Zahlen, die einen Real- und einen Imaginärteil (gekennzeichnet durch `j`) haben. Diese sind primär in wissenschaftlichen und technischen Berechnungen relevant.
    * Beispiel: `z = 2 + 3j`

#### Zeichenketten (`str`)

* **String (`str`):** Eine Sequenz von Zeichen (Buchstaben, Ziffern, Sonderzeichen), die in einfachen (`'`) oder doppelten (`"`) Anführungszeichen eingeschlossen wird. Strings werden für die Darstellung von Text verwendet.
    * Beispiele: `gruss = "Hallo Welt!"`, `name = 'Peter Pan'`, `fehlermeldung = "Error 404"`
    *

```
text = "Python ist toll!"
print(type(text))  # Ausgabe: <class 'str'>
```

* **f-Strings (Formatierte String-Literale):** Eine moderne und sehr lesbare Methode, um Variablenwerte und Ausdrücke direkt in Zeichenketten einzubetten. Man setzt ein `f` vor das öffnende Anführungszeichen und schreibt Variablen/Ausdrücke in geschweifte Klammern `{}`.
    *

```
benutzer_name = "Alex"
alter = 28
ausgabe = f"Benutzer {benutzer_name} ist {alter} Jahre alt."
print(ausgabe)  # Ausgabe: Benutzer Alex ist 28 Jahre alt.
```

#### Listen (`list`)

* **List (`list`):** Eine **geordnete** und **veränderliche (mutable)** Sammlung von Elementen. Listen können Elemente verschiedener Datentypen enthalten und werden durch eckige Klammern `[]` definiert.
    * Elemente haben einen Index (beginnend bei 0).
    * Man kann Elemente hinzufügen, entfernen und ändern.
    * Beispiele: `einkaufsliste = ["Äpfel", "Milch", "Brot"]`, `zahlen = [1, 2, 3, 4, 5]`, `gemischt = [10, "Hallo", True, 3.14]`
    *

```
meine_liste = [10, "Apfel", 20.5]
print(meine_liste[1])  # Ausgabe: Apfel (Zugriff auf das zweite Element)
meine_liste.append(False) # Fügt 'False' am Ende hinzu
print(meine_liste)     # Ausgabe: [10, 'Apfel', 20.5, False]
meine_liste[0] = 100   # Ändert das erste Element
print(meine_liste)     # Ausgabe: [100, 'Apfel', 20.5, False]
print(type(meine_liste)) # Ausgabe: <class 'list'>
```

#### Weitere nützliche Datentypen (Überblick)

* **Tupel (`tuple`):** Ähnlich wie Listen sind Tupel **geordnete** Sammlungen von Elementen. Der Hauptunterschied ist, dass Tupel **unveränderlich (immutable)** sind – einmal erstellt, können ihre Elemente nicht mehr geändert, hinzugefügt oder entfernt werden. Tupel werden mit runden Klammern `()` definiert.
    * Verwendung: Für Daten, die sich nicht ändern sollen (z.B. Farbdefinitionen RGB, Koordinaten).
    * Beispiel: `rgb_rot = (255, 0, 0)`, `koordinaten = (10.5, 49.3)`
    *

```
mein_tupel = (1, "zwei", 3.0)
print(mein_tupel[0])  # Ausgabe: 1
# mein_tupel[0] = 5  # Würde einen TypeError verursachen!
print(type(mein_tupel)) # Ausgabe: <class 'tuple'>
```

* **Dictionary (`dict`):** Eine **ungeordnete** (vor Python 3.7) bzw. **einfüge-geordnete** (ab Python 3.7) Sammlung von **Schlüssel-Wert-Paaren (key-value pairs)**. Jeder Schlüssel muss innerhalb eines Dictionaries eindeutig sein und wird verwendet, um auf den zugehörigen Wert zuzugreifen. Dictionaries sind **veränderlich** und werden mit geschweiften Klammern `{}` definiert.
    * Verwendung: Um zusammengehörige Informationen zu speichern und schnell über einen Schlüssel darauf zuzugreifen (z.B. Benutzereinstellungen, Telefonbuch).
    * Beispiel: `person = {"name": "Maria", "alter": 30, "stadt": "München"}`
    *

```
auto = {"marke": "VW", "modell": "Golf", "baujahr": 2022}
print(auto["modell"])  # Ausgabe: Golf
auto["farbe"] = "Blau" # Fügt ein neues Schlüssel-Wert-Paar hinzu
print(auto)          # Ausgabe: {'marke': 'VW', 'modell': 'Golf', 'baujahr': 2022, 'farbe': 'Blau'}
print(type(auto))    # Ausgabe: <class 'dict'>
```

### 2. Verzweigungen (Kontrollstrukturen: `if-elif-else`)

Verzweigungen sind fundamentale Kontrollstrukturen in der Programmierung. Sie ermöglichen es dem Programm, Entscheidungen zu treffen und unterschiedliche Code-Pfade auszuführen, basierend darauf, ob bestimmte **Bedingungen** (Ausdrücke, die zu `True` oder `False` evaluieren) erfüllt sind oder nicht. In Python ist die **Einrückung (Indentation)** des Codes nach einer `if`, `elif` oder `else` Anweisung entscheidend – sie definiert, welche Codezeilen zum jeweiligen Block gehören!

#### Einfache Verzweigung (`if-else`)
Die einfachste Form der Verzweigung. Wenn die Bedingung im `if`-Teil `True` ist, wird der Code-Block darunter ausgeführt. Andernfalls (wenn die Bedingung `False` ist) wird der Code-Block im optionalen `else`-Teil ausgeführt.

* Beispiel: Überprüfen, ob eine eingegebene Zahl positiv ist.

```
zahl_eingabe = input("Geben Sie eine Zahl ein: ")
zahl = int(zahl_eingabe)

if zahl > 0:
    print(f"Die Zahl {zahl} ist positiv.")
else:
    print(f"Die Zahl {zahl} ist Null oder negativ.")
```

#### Mehrfachverzweigung (`if-elif-else`)
Wenn es mehr als zwei mögliche Pfade gibt, verwendet man `elif` (Kurzform von "else if"). Python wertet die Bedingungen der Reihe nach von oben nach unten aus:
1.  Die `if`-Bedingung wird geprüft. Ist sie `True`, wird der zugehörige Block ausgeführt, und die restliche `elif`/`else`-Struktur wird übersprungen.
2.  Ist die `if`-Bedingung `False`, wird die erste `elif`-Bedingung geprüft. Ist diese `True`, wird ihr Block ausgeführt, und der Rest wird übersprungen.
3.  Dies setzt sich für alle weiteren `elif`-Blöcke fort.
4.  Ist keine der vorherigen Bedingungen `True`, wird der optionale `else`-Block am Ende ausgeführt.

* Beispiel: Umwandlung einer Schulnote in eine verbale Bewertung.

```
note = int(input("Geben Sie die Schulnote (1-6) ein: "))

if note == 1:
    print("Sehr gut!")
elif note == 2:
    print("Gut.")
elif note == 3:
    print("Befriedigend.")
elif note == 4:
    print("Ausreichend.")
elif note == 5:
    print("Mangelhaft.")
elif note == 6:
    print("Ungenügend.")
else:
    print("Ungültige Eingabe. Bitte eine Note zwischen 1 und 6 eingeben.")
```

* **Vergleich: Verschachteltes `if-else` vs. `if-elif-else`**
    Das ursprüngliche Beispiel zur Zahlenüberprüfung aus dem ersten Entwurf (hier leicht angepasst):

```
# Ursprünglicher Ansatz mit Verschachtelung
    # x = int(input("Geben Sie die erste Zahl ein: "))
    # y = int(input("Geben Sie die zweite Zahl ein: "))
    #
    # if x > y:
    #     print(f"{x} ist größer als {y}.")
    # else:  # x ist nicht größer als y, also x <= y
    #     if x == y:  # Verschachtelte Prüfung
    #         print(f"{x} und {y} sind gleich.")
    #     else:  # x < y
    #         print(f"{y} ist größer als {x}.")
```

Dieses kann mit `elif` oft flacher und damit lesbarer gestaltet werden:

```
x_neu = int(input("Geben Sie die erste Zahl ein: "))
y_neu = int(input("Geben Sie die zweite Zahl ein: "))

if x_neu > y_neu:
    print(f"{x_neu} ist größer als {y_neu}.")
elif y_neu > x_neu:  # Dies ersetzt die Verschachtelung klarer
    print(f"{y_neu} ist größer als {x_neu}.")
else:  # Der verbleibende Fall ist x_neu == y_neu
    print(f"{x_neu} und {y_neu} sind gleich.")
```

**Wichtige Punkte bei Verzweigungen:**
* Bedingungen verwenden oft Vergleichsoperatoren (`==`, `!=`, `>`, `<`, `>=`, `<=`) und logische Operatoren (`and`, `or`, `not`) aus Modul 1.
* Die Einrückung (üblicherweise 4 Leerzeichen) ist zwingend und Teil der Python-Syntax!
