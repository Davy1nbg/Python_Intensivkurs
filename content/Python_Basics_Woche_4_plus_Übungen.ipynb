### 1. Einleitung: Warum Funktionen und Module? (5 Min.)

Stellt euch vor, ihr schreibt ein längeres Programm, z.B. ein kleines Spiel oder eine Datenanalyse. Ohne Struktur wird der Code schnell unübersichtlich, lang und schwer zu warten. Bestimmte Code-Abschnitte wiederholen sich vielleicht oft.

* **Funktionen** sind wie kleine, benannte "Mini-Programme" innerhalb eures Hauptprogramms. Sie bündeln Code, der eine bestimmte Aufgabe erledigt. Das macht euren Code:
    * **Strukturierter:** Leichter zu lesen und zu verstehen.
    * **Wiederverwendbar:** Einmal geschrieben, könnt ihr eine Funktion beliebig oft aufrufen.
    * **Einfacher zu testen und zu debuggen:** Kleine, isolierte Einheiten sind leichter auf Fehler zu überprüfen.

* **Module** gehen noch einen Schritt weiter. Sie sind Dateien, die Python-Definitionen und -Anweisungen (also z.B. Funktionen, aber auch Klassen oder Variablen) enthalten. Das erlaubt:
    * **Organisation:** Code in logische Einheiten aufteilen (z.B. alle mathematischen Hilfsfunktionen in ein eigenes Modul).
    * **Wiederverwendung in verschiedenen Projekten:** Einmal geschriebene Module können in vielen Programmen genutzt werden.
    * **Nutzung von Standardfunktionalität:** Python kommt mit einer riesigen **Standardbibliothek** – einer Sammlung von Modulen für alle möglichen Aufgaben (Mathe, Zufallszahlen, Datum/Zeit, Webzugriff etc.).

### 2. Funktionen (insg. 45 Min.)

#### a) Was sind Funktionen? Definition und Aufruf (10 Min.)

Eine **Funktion** ist ein benannter Block von Code, der eine spezifische Aufgabe ausführt. Man "definiert" die Funktion einmal und kann sie dann beliebig oft "aufrufen".

**Definition einer Funktion:**
Das Schlüsselwort `def` leitet die Definition einer Funktion ein.

```
# Definition einer einfachen Funktion
def begruessung():
    print("Hallo Welt!")
    print("Willkommen zur Python-Schulung.")

# Das Programm macht hier noch nichts, die Funktion wurde nur "bekannt gemacht".
```

**Aufruf einer Funktion:**
Um den Code in der Funktion auszuführen, rufen wir sie bei ihrem Namen auf, gefolgt von Klammern `()`.

```
# Aufruf der Funktion
begruessung()
# Ausgabe:
# Hallo Welt!
# Willkommen zur Python-Schulung.

begruessung() # Kann beliebig oft aufgerufen werden
# Ausgabe:
# Hallo Welt!
# Willkommen zur Python-Schulung.
```

**Wichtig:** Eine Funktion muss definiert sein, *bevor* sie zum ersten Mal aufgerufen wird.

#### b) Parameter (Argumente) und Rückgabewerte (`return`) (15 Min.)

Funktionen werden noch nützlicher, wenn wir ihnen Daten übergeben (**Parameter**) und wenn sie uns Ergebnisse zurückliefern können (**Rückgabewerte**).

**Parameter:**
Parameter sind Variablen, die in der Funktionsdefinition innerhalb der Klammern `()` aufgelistet werden. Sie dienen dazu, Werte an die Funktion zu übergeben, mit denen sie arbeiten kann. Beim Aufruf der Funktion nennt man die übergebenen Werte **Argumente**.

```
def persoenliche_begruessung(name): # 'name' ist ein Parameter
    print(f"Hallo {name}!")
    print("Schön, dich hier zu haben.")

# Aufrufe mit verschiedenen Argumenten
persoenliche_begruessung("Anna")   # "Anna" ist das Argument für den Parameter 'name'
# Ausgabe:
# Hallo Anna!
# Schön, dich hier zu haben.

persoenliche_begruessung("Max")    # "Max" ist das Argument
# Ausgabe:
# Hallo Max!
# Schön, dich hier zu haben.
```

Eine Funktion kann mehrere Parameter haben:

```
def addiere(zahl1, zahl2):
    summe = zahl1 + zahl2
    print(f"Die Summe von {zahl1} und {zahl2} ist {summe}.")

addiere(5, 3)       # Ausgabe: Die Summe von 5 und 3 ist 8.
addiere(100, -20)   # Ausgabe: Die Summe von 100 und -20 ist 80.
```

**Rückgabewerte (`return`):**
Oft soll eine Funktion nicht nur etwas ausgeben, sondern ein Ergebnis berechnen und dieses an den aufrufenden Teil des Programms zurückgeben. Dafür verwendet man das Schlüsselwort `return`.

```
def multipliziere(a, b):
    produkt = a * b
    return produkt # Die Funktion gibt den Wert von 'produkt' zurück

# Den Rückgabewert in einer Variablen speichern
ergebnis = multipliziere(7, 6)
print(f"Das Ergebnis der Multiplikation ist: {ergebnis}") # Ausgabe: Das Ergebnis der Multiplikation ist: 42

# Rückgabewerte direkt verwenden
print(f"3 * 4 = {multipliziere(3, 4)}") # Ausgabe: 3 * 4 = 12

# Eine Funktion ohne 'return' oder mit einem leeren 'return' gibt implizit 'None' zurück.
def mach_etwas():
    x = 10
    # Kein return

wert = mach_etwas()
print(wert) # Ausgabe: None
```

Sobald `return` in einer Funktion ausgeführt wird, wird die Funktion sofort beendet, und der Wert wird zurückgegeben.

#### c) Geltungsbereich: Lokale vs. Globale Variablen (10 Min.)

Der **Geltungsbereich** (Scope) einer Variable bestimmt, wo im Code auf diese Variable zugegriffen werden kann.

* **Lokale Variablen:**
    * Werden *innerhalb* einer Funktion definiert.
    * Sind nur *innerhalb* dieser Funktion sichtbar und gültig.
    * "Sterben", wenn die Funktion beendet wird.
    * Helfen, Namenskonflikte zu vermeiden und Funktionen unabhängig voneinander zu halten.

```
def meine_funktion():
    lokale_var = "Ich bin lokal" # lokale_var ist lokal
    print(lokale_var)

meine_funktion()  # Ausgabe: Ich bin lokal
# print(lokale_var) # Fehler! NameError: name 'lokale_var' is not defined
                  # weil lokale_var außerhalb der Funktion nicht existiert.
```

* **Globale Variablen:**
    * Werden *außerhalb* aller Funktionen definiert (auf der obersten Ebene des Skripts).
    * Sind von überall im Code lesbar, sowohl außerhalb als auch innerhalb von Funktionen.

```
globale_var = "Ich bin global" # globale_var ist global

def andere_funktion():
    print(f"In der Funktion: {globale_var}") # Zugriff auf globale Variable (lesen)

andere_funktion()                 # Ausgabe: In der Funktion: Ich bin global
print(f"Außerhalb der Funktion: {globale_var}") # Ausgabe: Außerhalb der Funktion: Ich bin global
```

**Achtung beim Ändern globaler Variablen in Funktionen:**
Wenn man einer Variablen innerhalb einer Funktion einen Wert zuweist, wird standardmäßig eine *neue lokale* Variable mit diesem Namen erstellt. Will man die *globale* Variable ändern, muss man das Schlüsselwort `global` verwenden. **Dies sollte aber sparsam eingesetzt werden, da es die Lesbarkeit und Wartbarkeit des Codes erschweren kann!**

```
zaehler_global = 0

def erhoehe_zaehler_lokal():
    zaehler_global = 10 # Erstellt eine NEUE LOKALE Variable 'zaehler_global'
    print(f"Zähler in Funktion (lokal): {zaehler_global}") # Ausgabe: 10

def erhoehe_zaehler_global():
    global zaehler_global # Sagt Python: Wir meinen die globale Variable
    zaehler_global += 1
    print(f"Zähler in Funktion (global): {zaehler_global}")

erhoehe_zaehler_lokal()
print(f"Zähler außerhalb nach lokalem Aufruf: {zaehler_global}") # Ausgabe: 0 (globale wurde nicht geändert)

erhoehe_zaehler_global()
print(f"Zähler außerhalb nach globalem Aufruf: {zaehler_global}") # Ausgabe: 1
erhoehe_zaehler_global()
print(f"Zähler außerhalb nach weiterem globalem Aufruf: {zaehler_global}") # Ausgabe: 2
```

**Empfehlung:** Funktionen sollten idealerweise Daten über Parameter erhalten und Ergebnisse über `return` zurückgeben, anstatt direkt auf globale Variablen zuzugreifen oder diese zu modifizieren.

### 3. Module (insg. 35 Min.)

#### a) Was sind Module? Standardbibliotheken nutzen (15 Min.)

Ein **Modul** ist einfach eine Python-Datei (`.py`), die wiederverwendbaren Code enthält (Funktionen, Klassen, Variablen). Python selbst kommt mit einer großen Sammlung von Modulen, der **Standardbibliothek**. Diese Module bieten fertige Lösungen für häufige Aufgaben.

**Importieren von Modulen:**
Um die Funktionalität eines Moduls nutzen zu können, müssen wir es zuerst mit dem `import`-Befehl importieren.

**Beispiel: Das `math`-Modul**
Das `math`-Modul enthält mathematische Funktionen und Konstanten.

```
import math # Importiert das gesamte Modul 'math'

# Zugriff auf Funktionen/Konstanten im Modul über Modulname.Funktionsname
wurzel_von_16 = math.sqrt(16) # sqrt() für Quadratwurzel
print(f"Die Wurzel aus 16 ist: {wurzel_von_16}") # Ausgabe: 4.0

print(f"Der Wert von Pi ist: {math.pi}") # Konstante pi
print(f"2 hoch 5 ist: {math.pow(2, 5)}") # pow() für Potenzierung

# Man kann auch spezifische Teile eines Moduls importieren:
from math import sin, cos # Importiert nur sin und cos Funktionen

print(f"Sinus von 0.5: {sin(0.5)}") # Direkter Aufruf ohne 'math.' davor
# print(math.tan(0.5)) # Fehler! tan wurde nicht explizit importiert in dieser Form

# Oder alles aus einem Modul importieren (Vorsicht bei großen Modulen wg. Namenskonflikten!):
# from math import * # Nicht empfohlen für größere Projekte
# print(factorial(5)) # factorial() wäre jetzt direkt verfügbar
```

**Beispiel: Das `random`-Modul**
Das `random`-Modul wird verwendet, um Zufallszahlen zu erzeugen oder zufällige Auswahlen zu treffen.

```
import random

# Zufällige Ganzzahl zwischen a und b (inklusive)
zufallszahl = random.randint(1, 6) # Wie beim Würfeln
print(f"Zufallszahl zwischen 1 und 6: {zufallszahl}")

# Zufällige Fließkommazahl zwischen 0.0 und 1.0
zufalls_float = random.random()
print(f"Zufällige Fließkommazahl: {zufalls_float}")

# Ein zufälliges Element aus einer Liste auswählen
meine_liste = ["Apfel", "Banane", "Kirsche"]
zufalls_obst = random.choice(meine_liste)
print(f"Zufälliges Obst: {zufalls_obst}")

# Elemente einer Liste mischen (in-place, d.h. die Liste selbst wird verändert)
namen = ["Alice", "Bob", "Charlie", "David"]
print(f"Namen vorher: {namen}")
random.shuffle(namen)
print(f"Namen gemischt: {namen}")
```

#### b) Einführung: Eigene Module erstellen und verwenden (10 Min.)

Ihr könnt auch eure eigenen Module erstellen, um euren Code zu organisieren und wiederverwendbar zu machen.

**Schritt 1: Modul-Datei erstellen**
Erstellt eine neue Python-Datei (z.B. `meine_hilfsfunktionen.py`). In diese Datei schreibt ihr eure Funktionen (oder Klassen, Variablen), die ihr wiederverwenden wollt.

*Beispielinhalt für `meine_hilfsfunktionen.py`:*

```
# meine_hilfsfunktionen.py

PI_KONSTANTE = 3.1415926535

def kreis_flaeche(radius):
    return PI_KONSTANTE * radius * radius

def kreis_umfang(radius):
    return 2 * PI_KONSTANTE * radius

print("Modul 'meine_hilfsfunktionen' wurde geladen.") # Diese Zeile wird beim Import ausgeführt!
```

**Schritt 2: Eigenes Modul importieren**
In einem anderen Python-Skript (z.B. `hauptprogramm.py`), das sich **im selben Verzeichnis** wie `meine_hilfsfunktionen.py` befindet, könnt ihr euer Modul importieren:

*Beispielinhalt für `hauptprogramm.py`:*

```
# hauptprogramm.py

import meine_hilfsfunktionen # Importiert unsere eigene .py Datei

radius1 = 5
flaeche1 = meine_hilfsfunktionen.kreis_flaeche(radius1)
umfang1 = meine_hilfsfunktionen.kreis_umfang(radius1)

print(f"Kreis mit Radius {radius1}:")
print(f"  Fläche: {flaeche1}")
print(f"  Umfang: {umfang1}")

print(f"Unsere Pi-Konstante: {meine_hilfsfunktionen.PI_KONSTANTE}")

# Man kann auch hier spezifische Funktionen importieren:
# from meine_hilfsfunktionen import kreis_flaeche
# flaeche2 = kreis_flaeche(3)
# print(f"Fläche eines Kreises mit Radius 3: {flaeche2}")
```

Wenn ihr `hauptprogramm.py` ausführt, wird zuerst `meine_hilfsfunktionen.py` geladen (dabei wird auch das `print` im Modul ausgeführt), und dann könnt ihr auf dessen Inhalte zugreifen.

**Das `if __name__ == "__main__":` Konstrukt**
Manchmal möchte man, dass bestimmter Code in einer Moduldatei nur ausgeführt wird, wenn die Datei *direkt als Skript* gestartet wird, aber *nicht*, wenn sie als Modul importiert wird. Dafür gibt es eine spezielle Bedingung:

*In `meine_hilfsfunktionen.py` überarbeitet:*

```
# meine_hilfsfunktionen.py

PI_KONSTANTE = 3.1415926535

def kreis_flaeche(radius):
    return PI_KONSTANTE * radius * radius

def kreis_umfang(radius):
    return 2 * PI_KONSTANTE * radius

# Dieser Block wird nur ausgeführt, wenn meine_hilfsfunktionen.py direkt gestartet wird.
# Nicht, wenn es als Modul importiert wird.
if __name__ == "__main__":
    print("Modul 'meine_hilfsfunktionen' wurde direkt ausgeführt (Testcode).")
    r_test = 2
    print(f"Test - Fläche für Radius {r_test}: {kreis_flaeche(r_test)}")
    print(f"Test - Umfang für Radius {r_test}: {kreis_umfang(r_test)}")
```

Wenn ihr jetzt `hauptprogramm.py` ausführt, seht ihr "Modul 'meine_hilfsfunktionen' wurde direkt ausgeführt (Testcode)." *nicht* mehr. Dieser Teil dient oft dazu, Testcode für das Modul selbst bereitzustellen.

### 4. Zusammenfassung, Ausblick & Fragen (5 Min.)

**Was haben wir heute gelernt?**
* **Funktionen** (`def`) helfen uns, Code zu strukturieren, wiederholenden Code zu vermeiden und Aufgaben in logische Blöcke aufzuteilen.
* Mit **Parametern** können wir Daten an Funktionen übergeben.
* Mit `return` können Funktionen Ergebnisse zurückliefern.
* **Lokale Variablen** existieren nur innerhalb einer Funktion, **globale Variablen** überall (aber Vorsicht bei deren Modifikation!).
* **Module** (`import`) sind Python-Dateien mit Code, den wir wiederverwenden können.
* Python bietet eine umfangreiche **Standardbibliothek** mit vielen nützlichen Modulen (z.B. `math`, `random`).
* Wir können auch **eigene Module** erstellen, um unsere Projekte zu organisieren.

**Ausblick:**
Funktionen und Module sind fundamentale Bausteine in Python. In zukünftigen Einheiten werdet ihr sehen, wie sie zusammen mit anderen Konzepten (wie z.B. Objektorientierter Programmierung) verwendet werden, um noch komplexere und leistungsfähigere Anwendungen zu erstellen. Das Verständnis von Geltungsbereichen und der saubere Umgang mit Datenübergabe (Parameter/Return) sind entscheidend für guten Code!
