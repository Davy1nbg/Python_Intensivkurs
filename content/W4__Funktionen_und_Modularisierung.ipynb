### 1. Funktionen: Definition und Abstraktion

Funktionen bündeln logisch zusammengehörige Code-Abschnitte in benannte Einheiten. Dies erhöht die Lesbarkeit, erleichtert das Debugging und ermöglicht die Wiederverwendbarkeit von Code-Segmenten.

#### Syntax der Definition und des Aufrufs

Eine Funktion wird mit dem Schlüsselwort `def` deklariert. Der Codeblock wird erst ausgeführt, wenn die Funktion unter ihrem Namen aufgerufen wird.

```
# Definition
def begruessung():
    print("Willkommen zur Python-Schulung.")

# Aufruf
begruessung()
```

#### Datentransfer: Parameter und Rückgabewerte

* **Parameter:** Variablen in der Funktionsdefinition, die als Platzhalter für Eingabewerte dienen.
* **Rückgabewerte (`return`):** Ermöglichen es einer Funktion, ein berechnetes Ergebnis an den aufrufenden Kontext zurückzugeben. Sobald `return` erreicht wird, endet die Ausführung der Funktion unmittelbar.

```
def berechne_produkt(a, b):
    return a * b

ergebnis = berechne_produkt(7, 6)
print(f"Das Ergebnis ist: {ergebnis}")
```

---

### 2. Geltungsbereiche (Variable Scope)

Der Geltungsbereich definiert die Sichtbarkeit und Lebensdauer einer Variable.

* **Lokale Variablen:** Werden innerhalb einer Funktion definiert und sind nur dort zugänglich. Sie werden nach Beendigung der Funktion aus dem Speicher entfernt.
* **Globale Variablen:** Werden außerhalb von Funktionen definiert und sind im gesamten Skript lesbar.
* **Best Practice:** Funktionen sollten vorzugsweise über Parameter und Rückgabewerte kommunizieren, statt globale Variablen zu manipulieren.

---

### 3. Module und die Standardbibliothek

Ein Modul ist eine externe Datei (`.py`), die Definitionen und Anweisungen enthält.

#### Nutzung der Standardbibliothek

Python bietet integrierte Module für spezialisierte Aufgaben, die mittels `import` eingebunden werden:

* **`math`:** Mathematische Konstanten und Funktionen (z.B. `math.sqrt()`, `math.pi`).
* **`random`:** Erzeugung von Zufallswerten und stochastische Auswahl.

```
import math
import random

print(f"Wurzel aus 16: {math.sqrt(16)}")
print(f"Zufallszahl (1-6): {random.randint(1, 6)}")
```

#### Eigene Module und das `__main__`-Konstrukt

Eigene Dateien können ebenfalls als Module importiert werden. Um zu verhindern, dass Testcode beim Import ausgeführt wird, nutzt man die Bedingung `if __name__ == "__main__":`.

```
# In der Modul-Datei (z.B. hilfsfunktionen.py)
def utility():
    return "Nützliche Funktion"

if __name__ == "__main__":
    # Dieser Block läuft nur bei direktem Start der Datei
    print("Testlauf: " + utility())
```
